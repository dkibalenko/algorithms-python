### 1. Two Pointers, Time $O(n)$, Space $O(1)$

```python
def reverse(array: list[int]) -> list[int]:

    left = 0
    right = len(array) - 1

    while left < right:
        temp = array[left]  # temp reference of left marker's object, DO NOT lose reference to this inital object

        array[left] = array[right]  # overwrite left marker with right marker
        array[right] = temp  # 'temp' reference still points to left marker's object before it was overwritten in previous step

        left += 1  # increment left marker
        right -= 1  # decrement right marker
    return array
```

```python
reverse([1, 2, 3, 4, 5, 6, 7])
```

### 2. Two Pointers, Tuple Unpacking, Time $O(n)$, Space $O(1)$

```python
def reverse2(array: list[int]) -> list[int]:
    left, right = 0, len(array) - 1
    while left < right:
        array[left], array[right] = array[right], array[left]
        left += 1
        right -= 1
    return array
```

```python
reverse2([1, 2, 3, 4, 5, 6, 7])
```

### 3. 

```python
def reverse3(array: list[int]) -> list[int]:
    n = len(array)
    for i in range(n // 2):
        array[i], array[-i - 1] = array[-i - 1], array[i]
    return array
```

```python
reverse3([1, 2, 3, 4, 5, 6, 7])
```

### 4.

```python
def reverse4(array):
    """Slicing reversely"""
    array[:] = array[::-1]
    return array
```

```python
reverse4([1, 2, 3, 4, 5, 6, 7])
```

### 5.

```python
def reverse5(array: list[int]) -> list[int]:
    """Method .reverse()"""
    array.reverse()
    return array
```

```python
reverse5([1, 2, 3, 4, 5, 6, 7])
```
