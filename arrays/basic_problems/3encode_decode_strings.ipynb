## Encode and Decode Strigs

Design an algorithm to **encode a list of strings** to a **single string.** The encoded string is then **decoded back** to the **original list** of strings. Encode and decode needs to be **stateless** - don't store any extra data in a separate data structure.

Please implement `encode` and `decode`.

- `strs[i]` contains only UTF-8 characters.

```python
Input: ["neet","code","love","you"]
Output:["neet","code","love","you"]

Input: ["we","say",":","yes"]
Output: ["we","say",":","yes"]
```
Explanation:
One possible **encode method** is: `"lint:;code:;love:;you`

---

- The basic assumption is, in the input we could have **any** possible character in the inputs not just limited to lowercase `a-z`.
- The difficult part is, how do we create some kind of **delimiter** between each word? How do we know **when** one word ends and another one begins?
- The focus is on **how we delimit each string** in the input array, to know where each string is starting and ending.

What if we use a `#` sign for separating strings?

- e.g. ["neet", "code"] -> "needcode";
- encode is ok. but for decoding we need to remember wher words end and start;
- we need something to separate words (e.g. `"neet#code"`);
- when we come across our 'delimiter' we skip it, and start think about the next word;
- we need to ensure, our delimiter(`#`) never shows up in any words (e.g. `"co#de"` -> `"neet#co#de"`), this means, having only a single special character delimiter is not going to work;

#### Solution:
- use `int#string` format, e.g. `"4#neet"` - `4` specifies the **length of the word** followed by `#` to delimit words;
- read the number (`x`) before `#`, transform it into `int`;
- count the next `x` characters after `#` delimiter;
- this works, even there are other `#` chars or `int` in the word, since we only read a single number followed by a single `#`.

````python
# Encode
def encode(strs: list[str]) -> str:
    res = ""

    for s in strs:
        res += str(len(s)) + "#" + s  # encoded single string (e.g. '5#hello')

    return res
````

````python
encode(["hello", "world", ",", "I'm", "here!"])
````

  #### Decode
  - store resulting list in `result = []`
  0. `i` pointer: iterate through the entire input string with `i` pointer. go through all chars in input string `while i < len(string)`

  1. `j` second pointer: iterate to identify starting integer(find `#`) - stops AT `#`

  2. fetch integer portion from string. Length tells, how many followoing chars we need to read after `j`
  3. define current string BOUNDS & fetch string portion
  4. add the decdoed string to `result`
  5. move `i` pointer to next encoded string

````python
# Decode
def decode(string: str) -> list[str]:  # "5#hello5#world1#,3#I'm5#here!" - ["hello", "world", ",", "I'm", "here!"]
    result = []
    i = 0

    while i < len(string):  # 0
        j = i

        while string[j] != "#":  # 1
            j += 1
        
        length = int(string[i:j])  # 2 (e.g."5" -> 5)

        # 3
        idx_from = j + 1
        idx_to = j + 1 + length

        decoded_string = string[idx_from:idx_to]

        # 4
        result.append(decoded_string)

        # 5
        i = idx_to

    return result
````

````python
decode("5#hello5#world1#,3#I'm5#here!")
````

#### Why it works
- Scans digits until # → extracts the length
- Uses that length to slice the exact substring
- Moves pointer i forward correctly
- Works even if strings contain:
  - spaces
  - punctuation
  - `#`
  - unicode
  - empty strings
