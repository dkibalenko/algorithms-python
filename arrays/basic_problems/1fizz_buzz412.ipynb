### 412. Fizz Buzz

Given an integer `n`, return a string array answer (1-indexed) where:

- `answer[i] == "FizzBuzz"` if i is divisible by `3` and `5`.
- `answer[i] == "Fizz"` if i is divisible by `3`.
- `answer[i] == "Buzz"` if i is divisible by `5`.
- `answer[i] == i` (as a string) if none of the above conditions are true.

Example 1:
```shell
Input: n = 3
Output: ["1","2","Fizz"]
```

Example 2:
```shell
Input: n = 5
Output: ["1","2","Fizz","4","Buzz"]
```

Example 3
```shell
Input: n = 15
Output: ["1","2","Fizz","4","Buzz","Fizz","7","8","Fizz","Buzz","11","Fizz","13","14","FizzBuzz"]
```

````python
def fizz_buzz(n):
    answer = []
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            answer.append("FizzBuzz")
        elif i % 3 == 0:
            answer.append("Fizz")
        elif i % 5 == 0:
            answer.append("Buzz")
        else:
            answer.append(str(i))
    return answer
````

````python
fizz_buzz(5)
````

Given a list of integers:
1. Replace all integers that are **evenly divisable** by 3 with "fizz";
2. Replace all integers divisible by 5 with "buzz";
3. Replace all integers divisible by both 3 & 5 with "fizzbuzz"

Example:
```shell
numbers = [45, 22, 14, 65, 97, 72]
Output: ["fizbuzz", 22, 14, 97, "fizz"]
```

````python
def fizz_buzz_enum(numbers):
    for i, num in enumerate(numbers):
        if num % 5 == 0 and num % 3 == 0:
            numbers[i] = "fizzbuzz"
        elif num % 3 == 0:
            numbers[i] = "fizz"
        elif num % 5 == 0:
            numbers[i] = "buzz"
    return numbers
````

````python
numbers = [45, 22, 14, 65, 97, 72]

fizz_buzz_enum(numbers)
````
