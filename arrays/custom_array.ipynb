### Custmo List

```python
class MyArray:
    def __init__(self):
        self.length = 0
        self.data = {}

    def get_item(self, index):
        if index < 0 or index >= self.length:
            raise IndexError("Index out of range")
        return self.data[index]

    def append_item(self, item):
        self.data[self.length] = item
        self.length += 1
        return self

    def pop_item(self):
        if self.length == 0:
            raise IndexError("Pop from empty array")

        last_item = self.data[self.length - 1]
        del self.data[self.length - 1]
        self.length -= 1
        return last_item

    def delete_item(self, index):
        if index < 0 or index >= self.length:
            raise IndexError("Index out of range")

        item = self.data[index]

        for i in range(index, self.length - 1):
            self.data[i] = self.data[i + 1]

        del self.data[self.length - 1]
        self.length -= 1
        return item

    def insert_item(self, index, value):
        if index < 0 or index > self.length:
            raise IndexError("Index out of range")

        self.data[self.length] = None
        self.length += 1

        for i in range(self.length - 1, index, -1):
            self.data[i] = self.data[i - 1]

        self.data[index] = value

    def __str__(self):
        return (
            f"{__class__.__name__}, length: {self.length}, "
            f"data: {[self.data[i] for i in range(self.length)]}"
        )
```

### Pop item method

```python
class MyArray:
    def __init__(self):
        self.length = 0
        self.data = {}
    
    def append_item(self, item):
        self.data[self.length] = item
        self.length += 1
        return self

    def __str__(self):
        return (
            f"{__class__.__name__}, length: {self.length}, "
            f"data: {[self.data[i] for i in range(self.length)]}"
        )

    def pop_item(self):
        if self.length == 0:
            raise IndexError("Pop from empty array")

        last_item = self.data[self.length - 1]
        del self.data[self.length - 1]
        self.length -= 1
        return last_item
```

### Delete item method

```python
class MyArray:
    def __init__(self):
        self.length = 0
        self.data = {}

    def __str__(self):
        return (
            f"{__class__.__name__}, length: {self.length}, "
            f"data: {[self.data[i] for i in range(self.length)]}"
        )
    
    def append_item(self, item):
        self.data[self.length] = item
        self.length += 1
        return self

    def delete_item(self, index):
        # 1. ref item
        # 2. shift items left
        # 3. remove item from data
        # 4. decrement length
        if index < 0 or index >= self.length:
            raise IndexError("Index out of range")

        item = self.data[index]

        for i in range(index, self.length - 1):
            self.data[i] = self.data[i + 1]

        del self.data[self.length - 1]
        self.length -= 1
        return item
```

```python
array = MyArray()
array.append_item("a")
array.append_item("b")
array.append_item("c")
array.append_item("d")
print(array)
```

```python
array.delete_item(2)
print(array)
```

### Insert item method

```python
class MyArray:
    def __init__(self):
        self.length = 0
        self.data = {}

    def __str__(self):
        return (
            f"{__class__.__name__}, length: {self.length}, "
            f"data: {[self.data[i] for i in range(self.length)]}"
        )
    
    def insert_item(self, index, value):
        # 1. extend list
        # 2. increment length
        # 3. shift items right from the end to index
        # 4. insert item
        if index < 0 or index > self.length:
            raise IndexError("Index out of range")

        self.data[self.length] = None
        self.length += 1

        for i in range(self.length - 1, index, -1):
            self.data[i] = self.data[i - 1]

        self.data[index] = value
```

```python
array = MyArray()
array.insert_item(0, "a")
array.insert_item(1, "b")
print(array)
```
