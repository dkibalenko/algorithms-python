### 217. Contains Duplicate
Given an integer array `nums`, return `true` if any value appears **at least twice** in the array, and return `false` if every element is distinct.

**Example 1**:
```shell
Input: nums = [1,2,3,1]
Output: true
```
Explanation: The element 1 occurs at the indices 0 and 3.

**Example 2**:
```shell
Input: nums = [1,2,3,4]
Output: false
```
Explanation: All elements are distinct.

**Example 3**:
```shell
Input: nums = [1,1,1,3,3,4,3,2,4,2]
Output: true
```

**Constraints**:
```shell
1 <= nums.length <= 10^5
-109 <= nums[i] <= 109
```

[Visual](../images/217contains_duplicates.png)

````python
def func(nums: list[int]) -> bool:
    # hash set - store visited items
    # setup
    # seen = set()
    # scan input - for loop - full scan

    # 1. check if i in seen -> return True
    # 2. add i to seen
    # 3. return False

    seen = set()

    for i in range(len(nums)):
        if nums[i] in seen:
            return True
        seen.add(nums[i])

    return False
````

Hash set - a membership flag

````python
from typing import Callable
import time
from functools import wraps


def timer(rep: int) -> Callable:
    def decorator(func: Callable) -> Callable:
        @wraps(func)
        def wrapper(*args, **kwargs):
            start = time.perf_counter()
            for _ in range(rep):
                result = func(*args, **kwargs)
            end = time.perf_counter()
            print(f"{func.__name__} took {end - start:.6f} seconds")
            return result
        return wrapper
    return decorator
````

````python
@timer(5)
def contains_duplicates(nums: list[int]) -> bool:
    seen = set()

    for i in range(len(nums)):
        if nums[i] in seen:
            return True
        seen.add(nums[i])
    return False
````

````python
from collections import Counter


@timer(5)
def contains_dupl(nums):
    """Use Counter class from collections, Time O(n + n), Space O(n)."""

    counted = Counter(nums)  # O(n) build

    for item in counted.values():  # O(n)
        if item > 1:
            return True

    return False
````

### Sorting
Sorting first makes duplicates group together, be adjacent. We can use this property to evaluate the input for containing duplicates.

- Time $O(n\space log\space n)$ - total. Sorting + one pass $O(n)$ checking for duplicates
- Space $O(1)$ - sorting in-place no extra space taken

[Visual](../images/217contains_duplicates_sort.png)

````python
@timer(5)
def cont_dupl(nums: list[int]) -> bool:
    # sorting
    # setup
    # sort nums in-place
    # scan input - for loop [1, len(nums)]

    # 1. if i = i - 1 -> return True
    # 2. return False

    nums.sort()

    for i in range(1, len(nums)):
        if nums[i] == nums[i - 1]:
            return True
    return False
````
