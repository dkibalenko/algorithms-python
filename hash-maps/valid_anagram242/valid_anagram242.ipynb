### 242. Valid Anagram

[Neet](https://neetcode.io/problems/is-anagram/solution)

Given two strings `s` and `t`, return `true` if `t` is an anagram of `s`, and `false` otherwise.

Example 1:

```
Input: s = "anagram", t = "nagaram"
Output: true
```

Example 2:
```
Input: s = "rat", t = "car"
Output: false
```
**Constraints**:

- `1 <= s.length, t.length <= 5 * 104`
- `s` and `t` consist of lowercase English letters.

### 1. Sorting

Time & Space Complexity
- Time $O(n\space log\space n + m\space log\space m)$
- Space $O(1)$ or $O(n + m)$

Where $n$ is the length of string $s$ and $m$ is the length of string $t$.

````python
def isAnagram(self, s: str, t: str) -> bool:
    if len(s) != len(t):
        return False

    return sorted(s) == sorted(t)
````

### 2. Hash Map

Time & Space Complexity
- Time $O(n+m)$
- Space $O(1)$  since we have at most 26 different characters.

````python
def isAnagram(self, s: str, t: str) -> bool:
    if len(s) != len(t):
        return False

    countS, countT = {}, {}

    for i in range(len(s)):
        countS[s[i]] = 1 + countS.get(s[i], 0)
        countT[t[i]] = 1 + countT.get(t[i], 0)

    return countS == countT
````

### 3. Hash Table (Using Array)

Time & Space Complexity
- Time $O(n+m)$
- Space $O(1)$  since we have at most 26 different characters.

````python
def is_anagram(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False

    seen_s = {}  # {"a": 3, "n": 1, "g": 1, "r": 1, "m": 1}
    seen_t = {}

    # count chars
    for i in range(len(s)):
        seen_s[s[i]] = 1 + seen_s.get(s[i], 0)
        seen_t[t[i]] = 1 + seen_t.get(t[i], 0)

    # check num occurences in both strings
    for char in seen_s:
        if seen_s[char] != seen_t.get(char, 0):
            return False

    return True
````
