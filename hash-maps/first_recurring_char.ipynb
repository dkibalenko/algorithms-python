### First Recurring Char
Given an array. Return the first recurring element. If there is no recurring items, return `'undefined'`.
1. array = [2, 5, 1, 2, 3, 5, 1, 2, 4] -> 2
2. array = [2, 1, 1, 2, 3, 5, 1, 2, 4] -> 1
3. array = [2, 3, 4, 5] -> undefined

```python
def first_recurring_char(array):
    seen = set()

    for char in array:
        if char in seen:
            return char
        seen.add(char)
    return "undefined"
```

```python
def first_recurring_char2(array):
    bucket = {}

    for char in array:
        if char in bucket:
            return char
        else:
            bucket[char] = None

    return "undefined"
```

```python
def first_recurring_character3(input_list):
    seen = {}

    for i, value in enumerate(input_list):
        if value in seen:
            return value
        else:
            seen[value] = i
    return None
```
