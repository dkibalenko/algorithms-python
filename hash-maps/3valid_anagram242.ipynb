### 242. Valid Anagram
Given two strings `s` and `t`, return `true` if `t` is an anagram of `s`, and `false` otherwise.

**Example 1:**
```shell
Input: s = "anagram", t = "nagaram"
Output: true
```
**Example 2:**
```shell
Input: s = "rat", t = "car"
Output: false
```
**Constraints:**

- `1 <= s.length, t.length <= 5 * 10^4`
- `s` and `t` consist of lowercase English letters.

**Follow up:** What if the inputs contain Unicode characters? How would you adapt your solution to such a case?

- I need to test the second input string `t` and return `True` if it's an anagram of the first input string `s`, that is the **rearranging** all letters of `s` exaclty once for building `t` string, otherwise return `False`.

- **Type**: boolean

**Step 2 - examples + edge cases**

`s = "anagram", t = "nagaram"` should return `True`, since `t` is anagaram of `s`, it uses all the same chars exactly once and has the same length.

**Step 3 - brute force**

**Sorting**: If two strings are anagrams, they must contain **exactly the same characters** with the same frequencies.
By sorting both strings, all characters will be arranged in a consistent order.
If the two sorted strings are identical, then every character and its count match, which means the strings are anagrams.
```py
def isAnagram(self, s: str, t: str) -> bool:
    # exit early
    if len(s) != len(t):
        return False

    # sorted strings should match
    return sorted(s) == sorted(t)
```
- Time $O(n\space log\space n) + O(m\space log\space m)$
- Space $O(1)$ or $O(n + m)$ depending on the sorting algorithm

> Where `n` is the length of string `s` and `m` is the length string `t`

Python's `sorted()` and `list.sort()` both use Timsort — invented by Tim Peters in 2002, specifically for Python, now also used in Java, Swift, and Android.

**Step 4 - select algorithm + state complexity**

**Hash map counting**
- If two strings are anagrams, they must use the **same characters with the same frequencies**.
- Instead of sorting, we can **count** how many times **each character appears** in both strings.
- By using two hash maps, we **track the frequency** of every character in each string.
- If both frequency maps match exactly, then the strings contain the same characters with same frequencies, meaning they are anagrams.

Stop early when both input strings have different length, return `False`.
Build 2 **hash maps** of input strings char counts, mapping thier chars to counts. Then compare resulting hash maps of char counts and if they equal return `True` otherwise `False`.

- Time is $O(n + m)$ since we traversing `s` and `t` in parallel and compare hash maps with `==` which is $O(n)$ time
- Space is $O(1)$ constant, since we store maximum 26 chars of Englisgh lowercase letters.

> Use hash maps to store the input strings characters mapping to their counts and compare the resulting hash maps.

Hash map - a frequency count

````python
def valid_anagram(s: str, t: str) -> bool:  # "anagram", t = "nagaram"
    # hash map
    # setup
    # CORE FACT: return False if len(s) != len(t)
    # set two hash maps for s, t chars counts: s, t = {}, {}
    # scan s - for loop - full scan

    # 1. count s, t chars: add/increment char count for both strings in parallel
    # 2. compare resulting hash maps: if not equal - return False
    # 3. return True

    if len(s) != len(t):  # core detail
        return False

    seen_s, seen_t = {}, {}

    # 1. count chars
    for i in range(len(s)):
        seen_s[s[i]] = seen_s.get(s[i], 0) + 1
        seen_t[t[i]] = seen_t.get(t[i], 0) + 1

    # 2. compare hash maps
    # if seen_s != seen_t:
    #     return False
    for char in seen_s:
        if seen_s[char] != seen_t.get(char, 0):
            return False

    return True
````

#### `if len(s) != len(t): return False` - core detail
**Correctness dependency**. Look at your comparison loop — it only iterates over `seen_s`'s keys, checking each against `seen_t.get(char, 0)`. It never checks whether `seen_t` has a key that `seen_s` doesn't. Question: if you deleted the `if len(s) != len(t): return False` line but kept the comparison loop exactly as-is, would the function still correctly reject a case like `s="ab"`, `t="abc"`? Trace it and see. What does that tell you about whether the length check is just an "early exit optimization," or something the correctness of your loop actually depends on?
- take `s="ab"`, `t="abc"` (`t` longer than `s`) and skip the length check. Your counting loop is `for i in range(len(s))`, so it only ever touches `t[0]` and `t[1]` — `t[2]='c'` is never counted. `seen_t` ends up `{'a':1,'b':1}`, identical to `seen_s`. Your comparison loop finds a match and returns `True` — a wrong answer, silently, no crash.

- Now flip it: `s="abc"`, `t="ab"` (`t` shorter). What happens when the loop tries `t[2]`? That's a second, different failure mode - `IndexError`.

> So the length check isn't just "the first anagram condition" — it's **what keeps the counting loop from either crashing or silently producing false positives**.

#### What if inputs contain Unicode characters?
Does your hash map approach's $O(1)$ space claim still hold if the alphabet isn't capped at 26 lowercase letters?
- With unicode chars (297,334) in input storing in the hash maps their counts could add significant overhead and turn space into $O(n + m)$. If thinking about 297,334 chars I think I should think about generators to optimize space complexity, do not load massive datasets into memory but process them one at a time.

> You have two quantities — `n` (length of the input string) and `k` (number of _distinct_ characters that could possibly appear). Your hash map's size is bounded by `min(n, k)`, not by `k` alone. For lowercase English, `k=26` and `n` is typically much larger, so `min(n,26)=26` — a fixed constant, hence $O(1)$. For Unicode, `k` is huge (hundreds of thousands of code points), but here's the question: **can a string of length `n` ever contain more than `n` distinct characters, no matter how large the alphabet is?** If not, what does that tell you `min(n, k)` collapses to when `k ≫ n`?

(Your generator idea is solving a different problem, by the way — streaming input so you don't hold the whole string in memory at once. That's about not loading a big _input_, not about how many _distinct keys_ end up in your hash map. Worth keeping those two space concerns separate.)

- Obviously, a string of length n can never contain more than n distinct characters. As I understand, unicode chars are all unique. Unicode is a standard that assigns a unique code point to every character. That's the entire purpose of the standard. This tells that `k <= n`, like in the beginning with only 26 English lowercases - there could be k or n items in the hash tables, what is matter with this?

> **The relationship for Unicode**: `k` (alphabet size) doesn't become `≤ n`; it's the opposite: Unicode has ~1.1 million possible code points, while this problem's constraint caps `n` at 50,000. So `k` (≈1,000,000+) is _much bigger_ than `n` (≤50,000) here — meaning `n` is now the smaller of the two.

That's exactly why it matters: `min(n, k)` is what bounds your hash map's key count.

- **With 26 lowercase letters**: `k=26` is tiny compared to any realistic `n`, so `min(n,26)=26` — a fixed constant, independent of input size. That's genuine $O(1)$ space.
- **With Unicode**: `k` is enormous, bigger than `n` itself, so `min(n,k)=n` — the bound now scales _with the input_, not with a fixed alphabet size.

So state it plainly: what's the corrected space complexity for your hash map solution once you drop the "lowercase English only" assumption and allow Unicode input? And does that answer the problem's own follow-up question?
- Space complexity for input with Unicode chars is $O(n)$, up to defined constraint of this task - 50000 char. This problem defined the constraint and dealing with all Unicode chars in the input will likely bring us to n distinct chars that need to be counted and stored in the hash maps.

Correct — $O(n)$ space once you drop the fixed-alphabet assumption, bounded by the 50,000 constraint. That's the right answer to the problem's own follow-up, and the reasoning holds: `min(n, k)` collapses to `n` when the alphabet is no longer small and fixed.
