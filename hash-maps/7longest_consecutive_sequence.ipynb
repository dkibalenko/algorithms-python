### 128. Longest Consecutive Sequence
Given an unsorted array of integers `nums`, return **the length of the longest consecutive elements sequence**.

You must write an algorithm that runs in $O(n)$ time.

**Example 1**:
```shell
Input: nums = [100,4,200,1,3,2]
Output: 4
```
Explanation: The longest consecutive elements sequence is [1, 2, 3, 4]. Therefore its length is 4.
**Example 2**:
```shell
Input: nums = [0,3,7,2,5,8,4,6,0,1]
Output: 9
```
**Example 3**:
```shell
Input: nums = [1,0,1,2]
Output: 3
``` 

**Constraints**:
```shell
0 <= nums.length <= 10^5
-10^9 <= nums[i] <= 10^9
```

### Reasoning
#### Step 1 - Restate the problem
I am given an integer array, not sorted, and I have to return **the longest consecutive sequence** using only $O(n)$ algorithm.

- The output is an integer meaning the length of the longest consecutive sequence
- `nums` array length may be 0
- integer elements in `nums` might be negative, zeros, positive and duplicated

#### Step 2 - Work examples out with Edge cases

````python
# nums = [100,4,200,1,3,2] returns 4, as 1, 2, 3, 4 are the longest consecutive string
````

````python
# nums = [] returns 0
# nums = [1] returns 1
# nums = [1, 1, 1] return 1
# nums = [-1, -10, -4, -2] returns 2
````

#### Step 3 - Brute Force

**Sorting + counter** - relying on **adjacency** of items

[Visual](../images/128longest_consec_seq_brute.png)

````python
def longest_consec_seq_brute(nums: list[int]) -> int:
    # brute force
    # setup
    # return 0 if input nums is empty
    # sort input array
    # set maximal sequence track: max_seq = 1, since there is at least 1 item
    # set counter: count = 1, start count from the first item
    # scan sorted input - for loop - [1, len(nums)]

    # 1. skip duplicates
    # 2. test if the current item is consecutive one after the previous one:
    #   if yes: increment counter: count += 1
    #   if not: reset counter to 1
    # 3. on each iteration: update max_seq if it's less than counter
    # 4. return max_seq

    if not nums: return 0

    max_seq = 1
    count = 1
    nums.sort()  # O(nlogn)

    for i in range(1, len(nums)):  # O(n)
        if nums[i] == nums[i - 1]:
            continue

        if nums[i] - nums[i - 1] == 1:
            count += 1
        else:
            count = 1

        max_seq = max(max_seq, count)

    return max_seq
````

````python
# longest_consec_seq([1, 0, 1, 2])
longest_consec_seq_brute([100,4,200,1,3,2])
````

- **Time**: $O(n\space log\space n)$
- **Space**: $O(1)$

#### Step 4 - Select Algorithm

**The Start-check and Expansion** - Hash set for **deduplication** & **lookup**

(A distinct hash-set trick — sequence detection, not counting — genuinely new)

> Iterating the set instead of the raw list guarantees each run only ever gets expanded once.

#### Step 5, 6 - Signature + Pseudocode

[Visual](../images/128longest_consec_seq_hashset.png)

````python
def longest_consec_seq_set(nums: list[int]) -> int:
    # hash set + find Start + Expansion
    # setup
    # early stop if nums is empty
    # count = 0
    # max_seq = 1

    # 1. build hash set for deduplication of nums: unique_nums = {...}
    # 2. scan unique_nums to find the Start of the sequence: num - 1 IN unique_nums?
    #   no: means Start is found -> count = 1 -> Expand the sequence while the sequnce GOES ON: while num + count IN uniqu_nums: count += 1
    #   yes: means the current item IS NOT Start of the sequence and must be ignored, other items will handle it
    # 3. update max_seq when it's smaller then count
    # 4. stop early when max_seq == len(unique_nums), since the max possible sequence is found
    # 5. return max_seq
    pass
````

#### Step 7 - Write Code

````python
def longest_consec_seq_set(nums: list[int]) -> int:
    if not nums: return 0

    count = 0
    max_seq = 1

    unique_nums = {item for item in nums}  # eliminate duplicates - O(n)

    for num in unique_nums:  # O(n)
        if not num - 1 in unique_nums:  # found Start of seq
            count = 1
            while num + count in unique_nums:  # expand seq (consecutive one exists)
                count += 1
        max_seq = max(max_seq, count)

        if max_seq >= len(unique_nums):
            break
    return max_seq
````

````python
# longest_consec_seq_set([100, 4, 200, 1, 3, 2])
longest_consec_seq_set([1, 1, 1, 1, 5, 3, 4, 2])
````

- **Time**: $O(n)$
- **Space**: $O(n)$

Iterating the set means each unique value gets checked as a start exactly once, and since every number belongs to at most one run, the total work across all expansions is bounded by `n`. That's genuine $O(n)$ time, $O(n)$ space now.

The early-break is a valid bonus: `max_seq` can never exceed the total count of distinct values (a run of consecutive distinct numbers can't be longer than the pool of distinct numbers itself), so once you hit that ceiling, no remaining unique value could possibly produce a longer run — safe to stop.

This closes the problem: brute force sort $O(n log n)$ → hash-set expansion-from-run-starts $O(n)$, with the recurring lesson (again) being to identify what you're recomputing before optimizing.

> Pattern: **hash set + expand only from true run starts** — file it for "check neighbors of a value in $O(1)$ without needing order" problems generally, not just consecutive integers.
