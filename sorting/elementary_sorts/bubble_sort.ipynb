### 🔥 Core Idea of Bubble Sort
Bubble sort **repeatedly walks through the list** and **swaps adjacent elements** that are out of order.  
On each full pass, the **largest** remaining element “bubbles up” to its correct position **at the end** of the list.

That’s the entire idea.

---

### 🧠 How it works conceptually
[Visualization link](https://visualgo.net/en/sorting)

```py
nums = [5, 1, 4, 2, 8]
```
1. **Outer loop**: makes exactly `len(nums)` passes
2. **Inner loop** `len(nums) - 1`
3. **CONDITTION** 2 adjacent elements(`j` & `j + 1`) & swaps them if needed

Bubble sort compares **adjacent** pairs:
- Compare 5 & 1 → swap
- Compare 5 & 4 → swap
- Compare 5 & 2 → swap
- Compare 5 & 8 → no swap

After one pass, the largest element (8) is at the end.

Next pass: bubble up the next largest.

You keep doing this **until no swaps** are needed.

---

### 🎯 Key properties
✔ Adjacent comparisons
- Only compares `nums[j]` and `nums[j+1]`.

✔ Swaps when needed
- If they’re in the wrong order → swap.

✔ **Largest** element moves to the end **each pass**
- Like a bubble rising to the surface.

✔ Stops early if no swaps
- **Optimized version** — if a **full pass has zero swaps**, the list is already sorted.

---

### 📉 Time & Space Complexity
- Worst case: $O(n²)$
- Average: $O(n²)$
- Best case (optimized): $O(n)$
- Space: $O(1)$ — in‑place

Bubble sort is simple but **slow**. It’s mainly **used for teaching**, not production.

---

### Brute‑Force Bubble Sort

````python
numbers = [99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0]
````

````python
def bubble_sort_bruteforce1(nums):
    """Time O(n^2), Space O(1)"""

    # repeat passes through the list
    for _ in range(len(nums)):

        for j in range(len(nums) - 1):
            # compare adjacent pairs
            if nums[j] > nums[j + 1]:
                temp = nums[j]
                nums[j] = nums[j + 1]
                nums[j + 1] = temp

    return nums
````

````python
bubble_sort_bruteforce1(numbers)
````

````python
# tuple unpacking technique
def bubble_sort_bruteforce2(nums):
    """Time O(n^2), Space O(1)"""

    for _ in range(len(nums)):
        for j in range(len(nums) - 1):
            if nums[j] > nums[j + 1]:
                nums[j], nums[j + 1] = nums[j + 1], nums[j]

    return nums
````

````python
bubble_sort_bruteforce2(numbers)
````

#### The Problem
This version always performs **all passes**, even if the list becomes sorted early.
- Always does `n` passes
- Inner loop always compares **all adjacent pairs**
- No early stopping
- Worst, average, and best case are **all** $O(n²)$

### Optimized Bubble Sort (early exit)
Stops early if **no swaps occur in a pass** — meaning the list is already sorted.

````python
def bubble_sort(nums: list[int]) -> list[int]:  # [99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0]

    # repeat passes through the list
    for i in range(len(nums)):
        swapped = False

        # traverse left->right, stop at stop-i
        for j in range(len(nums) - 1 - i):  # shrink unsorted region
            if nums[j] > nums[j + 1]:
                nums[j], nums[j + 1] = nums[j + 1], nums[j]
                swapped = True
        
        # stop early if already sorted
        if not swapped:
            break

    return nums
````

````python
bubble_sort([99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0])
````

- **Shrink Inner Loop**: after each pass, we know the **last part of the list is already sorted**, so we shrink the inner loop accordingly (`- i`):
```py
for j in range(len(nums) - 1 - i)
```
- **Stop Early**: track any swaps happened during a pass (`swapped = True`). If a full pass makes **zero swaps**, the list is already sorted, so we can **stop early**. These two optimizations eliminate unnecessary comparisons and give Bubble Sort a best‑case time of $O(n)$ instead of $O(n²)$.

````python
# traverse right->left, stop at i+1 (indices 0..i-1 already sorted)
def bub_sort(nums: list[int]) -> list[int]:
    # [99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0]
    for i in range(len(nums)):
        swapped = False

        for j in range(len(nums) - 1, i, -1):
            if nums[j - 1] > nums[j]:
                nums[j], nums[j - 1] = nums[j - 1], nums[j]
                swapped = True

        if not swapped:
            break
    return nums
````

````python
bub_sort([99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0])
````
