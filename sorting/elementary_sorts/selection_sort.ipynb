### Core Idea of Selection Sort

Selection Sort **repeatedly selects the smallest** remaining element and places it into its correct position **at the front** of the list.

Selection Sort selects the smallest element and moves it to the beginning.

### Core Rules
[Visualization link](https://visualgo.net/en/sorting)

1. **Outer loop**: makes exactly `len(nums)` passes. Initiates current min value - `min_idx = i`
- Divide the list into two regions: Sorted (left side) & Unsorted (right side)

2. **Inner loop** (`i + 1, len(nums)`): **scans unsorted** region for new smallest 
- `i + 1` shrinks the unsorted portion

3. Swap the smallest item with the item at position `i`

Sorted region grows from the left

- After pass `0` → index `0` is correct
- After pass `1` → indices `0–1` are correct
- After pass `i` → indices `0–i` are correct

Selection Sort always makes exactly `n–1` swaps

- This is a key property
- Even if the list is already sorted

Time complexity is always $O(n²)$

- Because you must scan the unsorted region **every time**

````python
def selection_sort(nums):
    # [99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0]
    for i in range(len(nums)):
        min_idx = i  # 1. set smallest index

        # 2. scan unsorted region to find new min
        for j in range(i + 1, len(nums)):
            if nums[j] < nums[min_idx]:
                min_idx = j

        # 3. swap new min into idx 'i'
        nums[i], nums[min_idx] = nums[min_idx], nums[i]

    return nums
````

````python
selection_sort([99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0])
````

- For each index `i`, refereced as **smallest**(`min_idx`), it scans the entire unordered list
- Finds the smallest element
- Swaps it into place

✔ Complexity
- Always $O(n²)$
- Always `n–1` swaps

### Grokking Algs

````python
def find_smallest(arr):
    smallest = arr[0]
    smallest_index = 0

    for i in range(1, len(arr)):
        if arr[i] < smallest:
            smallest = arr[i]
            smallest_index = i
    return smallest_index


def sel_sort(arr):
    new_arr = []
    for i in range(len(arr)):
        smallest = find_smallest(arr)
        new_arr.append(arr.pop(smallest))
    return new_arr
````

````python
sel_sort([5, 3, 6, 2, 10])
````

  public static void sort(Comparable[] a)
  {  // Sort a[] into increasing order
    int N = a.length;  // array length
    for (int i = 0; i < N; i++)
    {  // Exchange a[i] with smallest entry in a[i+1...N)
      int min = i;  // index of minimal entr
      for (int j = i+1; j < N; j++)
        if (less(a[j], a[min])) min = j;
      exch(a, i, min)
    }
  }
}
