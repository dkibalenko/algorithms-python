### Usage

The Insertion sort is used for sorting small datasets or nearly sorted datasets.

### 🎯 Core Idea of Insertion Sort

[Visualization link](https://visualgo.net/en/sorting)

Insertion Sort builds the sorted list **one element at a time** by inserting **each new** element into its correct position **among the already‑sorted** elements on its left.

1. **Outer loop(`for`)**: makes exactly `range(1, len(nums))` passes. Consider the first item already sorted
2. Set 2 pointers at the same index (`1`)
3. **Inner loop(`while`)**: scan **backwards** up to the first element (`j > 0`) & compare **current & one to left** items
4. Swap smaller to the left

````python
def insertion_sort_naive(nums):
    for i in range(1, len(nums)):
        j = i
        # scan backwards until the element is in the right place
        while j > 0 and nums[j] < nums[j - 1]:
            nums[j], nums[j - 1] = nums[j - 1], nums[j]
            j -= 1

    return nums
````

- Moves the element left by **swapping repeatedly**
- Works, but performs many swaps
- Still $O(n²)$

### Optimized version

### 🧠 How It Works (Step‑by‑Step)
1. Treat the first element as already sorted:
```py
for i in range(1, n):
```
2. Hold the next element (`key`): `key = nums[i]`
3. Compare it with elements to the left:
```py
j = i - 1

while j >= 0 and key < nums[j]:
```
4. Shift larger elements one position to the right:
```py
nums[j + 1] = nums[j]
j -= 1
```
5. Insert the `key` into the correct spot: `nums[j + 1] = key`
6. Repeat for all elements

The sorted region grows from **left to right**.

This is the canonical optimized version.

---

### ⏱ Time & Space Complexity
Time Complexity
- Best case: $O(n)$
  - (when the list is already sorted — only 1 comparison per element)
- Average case: $O(n²)$
- Worst case: $O(n²)$
  - (reverse‑sorted list — maximum shifting)

- Space Complexity
  - $O(1)$ — in‑place, no extra memory

---

````python
numbers = [99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0]
````

````python
def insertion_sort(nums: list[int]) -> list[int]:  # [99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0]
    # start from index 1
    for i in range(1, len(nums)):  # first pointer 'i' at current index
        key = nums[i]  # get current value to sort
        j = i - 1  # second pointer - one to left of current index

        # shift items to right, find place to insert
        while j >= 0 and key < nums[j]:  # compare current value & one to left
            nums[j + 1] = nums[j]  # move bigger to right
            j -= 1

        nums[j + 1] = key  # insertion
    
    return nums
````

````python
insertion_sort(numbers)
````

### Java Implementation

```java
public class Insertion
{
  public static void sort(Comparable[] a)
  { // Sort a[] into increasing order.
    int N = a.length;
    for (int i = 1; i < N; i++)
    { // Insert a[i] among a[i-1], a[i-2], a[i-3]... ..
      for (int j = i; j > 0 && less(a[j], a[j-1]); j--)
        exch(a, j, j-1);
    }
  }
  // See page 245 for less(), exch(), isSorted(), and main().
}
```
For each `i` from `0` to `N-1`, exchange a`[i]` with the entries that are smaller in `a[0]` through `a[i-1]`. As
the index `i` travels from left to right, the entries to its left are in sorted order in the array, so the array
is fully sorted when `i` reaches the right end.
