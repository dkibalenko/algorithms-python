### Timsort

Python's `sorted()` and `list.sort()` both use **Timsort** — invented by Tim Peters in 2002, specifically for Python, now also used in Java, Swift, and Android.

---

### The Core Idea

> **Real-world data is rarely random. It usually has partially sorted chunks already. Timsort exploits those chunks instead of ignoring them.**

Those naturally ordered chunks are called **runs**.
```shell
Random data (what most algorithms assume):
[5, 2, 8, 1, 9, 3, 7, 4, 6]

Real-world data (what Timsort expects):
[1, 3, 5, 8, | 2, 4, 6, | 7, 9]
  already sorted  already sorted  already sorted
  ↑ run 1         ↑ run 2         ↑ run 3
```

---

### How It Works
```shell
STEP 1 — Find or build runs
  Scan the array for naturally ascending or descending sequences.
  If a run is too short (< minrun, usually 32-64 elements)
  → extend it using Insertion Sort (fast on small + nearly sorted data)

STEP 2 — Merge runs
  Merge the runs together using a strategy borrowed from Merge Sort.
  Runs are pushed onto a stack and merged when size ratios are right
  (to keep the merge tree balanced).

  run1: [1, 3, 5, 8]
  run2: [2, 4, 6]       → merge → [1, 2, 3, 4, 5, 6, 8]
  run3: [7, 9]          → merge → [1, 2, 3, 4, 5, 6, 7, 8, 9]
```

---

### Complexity
```shell
Time              Space
Best case:    O(n)          ← already sorted, one run, no merging
Average case: O(n log n)
Worst case:   O(n log n)
Space:        O(n)          ← merge needs temporary buffer
```
> The $O(n) $best case is the practical payoff — a nearly sorted list
costs almost nothing to sort. No other general-purpose algorithm
guarantees this.

---

### Why It Matters for Interviews
1. "Is Python's sort stable?"
- YES. Timsort is stable — equal elements preserve their original relative order. This matters when sorting objects by multiple keys sequentially.

2. "What's the time complexity of `sorted()`?"
- `O(n log n)` worst and average. $O(n)$ best case.

3. "Why is `list.sort()` faster than `sorted()`?"
- `list.sort()` sorts in place — no new list created.
- `sorted()` creates a new list — $O(n)$ extra allocation.
- Same algorithm, different memory behaviour.

4. "When is insertion sort used inside Timsort?"
- On small runs (< ~64 elements). Insertion sort has low overhead and excellent cache performance on small nearly-sorted sequences — faster than merge sort at that scale despite $O(n²)$ worst case.
