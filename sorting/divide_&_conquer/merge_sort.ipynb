### Merge sort

Merge Sort is one of the most **efficient** and **stable** sorting algorithms based on the Divide and Conquer technique. It divides an input array into two halves, recursively sorts them, and then merges the two sorted halves using a function called `merge()`.

[Stable & Unstable algorithms](https://stackoverflow.com/questions/1517793/what-is-stability-in-sorting-algorithms-and-why-is-it-important)

### How Merge Sort Works

1. **Split** the unsorted list **into two halves** until each subarray has only **one element**.
2. Recursively sort each half using merge sort (they’re already sorted if they contain one element).
3. Merge the two sorted halves into a **single sorted list**.

Merge sort needs **additional memory** for temporary arrays during merging. Its space complexity is $O(n)$. That’s the tradeoff for guaranteed **performance** and **stability**.

This process continues until the list is reduced to single-element lists, which are naturally sorted. At that point, you’ve reached the base case, and the algorithm can start combining results.

In this example, `array` is the input list, and each recursive call keeps splitting the data into smaller sub-problems until there’s almost nothing left to split.

```python
def merge_sort(nums: list[int]):
    if len(nums) <= 1:
        return nums

    mid = len(nums) // 2

    # split
    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])

    # sort & merge
    res = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            res.append(left[i])
            i += 1
        else:
            res.append(right[j])
            j += 1
    
    res.extend(left[i:])
    res.extend(right[j:])

    return res
```

```python
merge_sort([20, 18, 1, 40, 4949, 8, 45, 0, 10])
```

```python
def merge_sort(array: list[int]) -> list[int]:
    n = len(array)

    # Base case: a list of length 0 or 1 is already sorted
    if n <= 1:
        return array

    mid = n // 2

    # Recursive case: split the array into two halves
    left_half = merge_sort(array[:mid])  # sort left half recursively
    right_half = merge_sort(array[mid:])  # sort right half recursively

    # Merge two sorted halves into a single sorted list
    return merge(left_half, right_half)


def merge(left: list[int], right: list[int]) -> list[int]:
    sorted_list = []  # result container
    i = j = 0  # pointers for left and right lists

    # Walk through both lists, picking the smaller current element each time
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_list.append(left[i])  # take from left
            i += 1
        else:
            sorted_list.append(right[j])  # take from right
            j += 1

    # At this point, at least one list is exhausted.
    # Append any remaining elements from left or right.
    sorted_list.extend(left[i:])  # leftover from left (if any)
    sorted_list.extend(right[j:])  # leftover from right (if any)

    return sorted_list
```

```python
numbers = [34, 7, 23, 32, 5, 62]
merge_sort([20, 18, 1, 40, 4949, 8, 45, 0, 10])
```

### What `merge` is doing conceptually
- You have two sorted lists: `left` and `right`.
- You maintain two pointers `i` and `j`.
- At each step, you compare `left[i]` and `right[j]` and append the smaller one.
- When one list is exhausted, you append the remainder of the other list.

This guarantees the result is sorted because you always take the smallest available next element.

---

### Time and space complexity
- Time:
  - Splitting: $O(log\space n)$ levels of recursion
  - Merging at each level: $O(n)$
  - Total: $O(n\space log\space n)$

- Space:
  - Needs extra space for merged lists → $O(n)$
