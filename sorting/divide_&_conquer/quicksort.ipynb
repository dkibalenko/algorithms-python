### Core Idea of Quicksort
Quicksort implementations use **median-of-three** (pick the median of first, middle, last elements as pivot) or **random pivot**. Both make it statistically impossible to consistently hit worst case.

- Pick a pivot
- Partition the array into elements less than the pivot and greater than the pivot
- Recursively sort the partitions.

### Three‑way partitioning approach

1. Pick a pivot
2. Split the array into:
    - elements less than pivot
    - elements equal to pivot
    - elements greater than pivot
3. Recursively sort the smaller and larger partitions
4. Concatenate the results

This avoids in‑place swapping and makes the logic crystal clear.

````python
def quick_sort(arr):
    """
    Time O(N log N) avg, O(N) worst if the pivot is always the smallest
    or largest element in the array.
    
    """
    # 1. Base case: arrays of length 0 or 1 are already sorted
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]  # choose the mid element as pivot

    # Partition into three lists:
    smaller = [num for num in arr if num < pivot]  # goes to left partition
    equal = [num for num in arr if num == pivot]   # stays with pivot group
    larger = [num for num in arr if num > pivot]   # goes to right partition

    # Recursively sort left and right partitions
    # Then concatenate: sorted smaller + equal + sorted larger
    return quick_sort(smaller) + equal + quick_sort(larger)
````

````python
numbers = [99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0]

quick_sort(numbers)
````

**Time complexity**

The time complexity of quick sort is determined by the **number of recursive calls** that are made.
- Each recursive call processes approximately $n/2$ elements, so the **total number of recursive calls** is approximately $log(n)$ - array halves each time.
- The number of calls doubles every level — but each call handles **half the data**. The doubling and halving **cancel out**, leaving exactly $O(n)$ total work per level. The running time of each recursive call is $O(n)$, so the total running time is $n\space log\space n$.
```shell
Array: [1, 2, 3, 4, 5, 6, 7, 8]   (n = 8)

Level 0:  [1, 2, 3, 4, 5, 6, 7, 8]          → scan 8 elements
               /              \
Level 1:  [1, 2, 3, 4]    [6, 7, 8]          → scan 4 + 4 = 8 elements
            /      \        /     \
Level 2: [1,2]   [4]   [6,7]   [8]            → scan 2+1+2+1 = 8 elements (ish)
          ...                                  → scan ~8 elements

Total work per level: O(n)
Number of levels:     O(log n)   ← array halves each time
Total:                O(n log n)
```
**Why Worst Case is $O(n²)$**
When the pivot is always the smallest (or largest) element — sorted input, pivot always min/max, $n$ levels:
```shell
[1, 2, 3, 4, 5]  pivot = 5
        ↓ 
smaller=[1,2,3,4]  equal=[5]  larger=[]

[1, 2, 3, 4]  pivot = 4
        ↓ 
smaller=[1,2,3]  equal=[4]  larger=[]
```
No halving. Each level removes exactly one element. That's $n$ levels of $O(n)$ work = $O(n²)$.

**Space Complexity**
```shell
Average: O(n log n) — each level creates new arrays totaling n elements,
                      log n levels × n elements = n log n
Worst:   O(n²)      — n levels × n elements per level
```

### Quicksort Using Lomuto In-Place Partition
**Space complexity**
```shell
Time:  O(n log n) avg — same logic
       O(n²) worst    — same trigger
Space: O(log n) avg   — call stack depth only
       O(n) worst     — call stack grows to n levels
```

````python
def quicksort(arr, low, high):
    # Base case: a segment of length 0 or 1 is already sorted
    if low < high:
        # Partition the array and get the pivot's final index
        pivot_index = lomuto_partition(arr, low, high)

        # Recursively sort elements before and after the pivot
        quicksort(arr, low, pivot_index - 1)
        quicksort(arr, pivot_index + 1, high)


def lomuto_partition(arr, low, high):
    pivot = arr[high]        # choose the last element as pivot
    i = low - 1              # boundary for elements < pivot

    # Scan through the segment and move smaller elements left
    for j in range(low, high):
        if arr[j] < pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]  # swap into the < pivot region

    # Place pivot in its correct sorted position
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1             # return pivot's final index
````

````python
numbers = [99, 44, 6, 2, 1, 5, 63, 87, 283, 4, 0]
id(numbers)
````

````python
quicksort(numbers, 0, len(numbers) - 1)
print(numbers)
id(numbers)
````

### 🎯 Core Idea of Lomuto Partition
1. Choose the **last** element as pivot  
Simple, predictable, easy to implement.

2. Maintain a boundary `i` for elements < pivot  
Everything left of `i` is guaranteed to be **smaller** than the pivot.

3. Scan with `j`  
When you find an element smaller than the pivot, expand the boundary and swap.

4. After scanning, put the pivot right after the `<` pivot region  
This places the pivot in its final sorted position.

4. Recursively sort left and right partitions  
No merging needed — the array is sorted **in place**.

````python
def two_sum(nums: list[int], target: int) -> list[int]:
    # [2, 7, 11, 15], target = 9
    left = 0
    right = len(nums) - 1

    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return []
````

````python
print(two_sum([3,2,4], 6))
````
