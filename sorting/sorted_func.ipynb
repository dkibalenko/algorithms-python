### 🔥 Core Idea of sorted()
`sorted()` takes any **iterable** and returns a **new sorted list**, using Python’s highly optimized **Timsort** algorithm.

It is:
- **Fast** ($O(n\space log\space n)$ worst‑case, $O(n)$ best‑case)
- **Stable** (**preserves order** of equal elements)
- **Flexible** (custom sorting with `key=` and `reverse=`)
- **Universal** (works on lists, tuples, sets, generators, files, dicts, etc.)

---

### 🧠 Why `sorted()` is so powerful

#### 1. It uses Timsort — one of the best practical sorting algorithms
Timsort is a hybrid of **merge sort** + **insertion sort**, optimized for real‑world data patterns like:
- partially sorted lists
- repeated values
- runs of increasing/decreasing order

This gives it:
- $O(n)$ best case
- $O(n\space log\space n)$ worst case
- **Excellent performance on nearly‑sorted data**

#### 2. It works on ANY iterable
Examples:
```py
sorted([3, 1, 2])
sorted((3, 1, 2))
sorted({3, 1, 2})
sorted("python")
sorted(open("file.txt"))
sorted(my_generator())
```

#### 3. It never mutates the original
Unlike `.sort()`, which sorts a list **in place**, `sorted()` always returns a **new list**.

This makes it:
- safer
- more functional
- easier to reason about

#### 4. It supports custom sorting with `key=`
This is where the **real power** is.

Examples:
```py
sorted(words, key=len)               # by length
sorted(users, key=lambda u: u.age)   # by attribute
sorted(files, key=os.path.getsize)   # by file size
sorted(data, key=str.lower)          # case-insensitive
```

#### 5. It supports reverse sorting
```py
sorted(nums, reverse=True)
```

---

### 📌 Best Use Cases (the ones you should remember)
#### 1. Sorting **complex objects**
```py
sorted(people, key=lambda p: p.last_name)
```

#### 2. Sorting **without mutating** the original
```py
new_list = sorted(old_list)
```

#### 3. Sorting by **multiple criteria**
```py
sorted(users, key=lambda u: (u.age, u.last_name))
```

#### 4. Sorting **text case‑insensitively**
```py
sorted(words, key=str.lower)
```

#### 5. Sorting **dictionary items**
```py
sorted(my_dict.items(), key=lambda item: item[1])  # by value
```

#### 6. Sorting **streams, generators**, or **file lines**
```py
sorted(line.strip() for line in open("log.txt"))
```

#### 7. Sorting **with custom rules**
```py
priority = {"high": 0, "medium": 1, "low": 2}
sorted(tasks, key=lambda t: priority[t.level])
```

### ⚡ `sorted()` vs `.sort()` — when to use which
| Feature                              | `sorted()` | `.sort()`      |
| ------------------------------------ | ---------- | -------------- |
| Returns new list                     | ✔          | ✘              |
| Mutates original                     | ✘          | ✔              |
| Works on any iterable                | ✔          | ✘ (lists only) |
| Functional style                     | ✔          | ✘              |
| Slightly slower (allocates new list) | ✔          | —              |
| Best for pipelines, generators       | ✔          | —              |

Rule of thumb:  
- Use `sorted()` unless you must **mutate the list in place**.

````python
# Sorting a list of dictionaries by multiple fields:

users = [
    {"name": "Alice", "age": 30},
    {"name": "Bob",   "age": 25},
    {"name": "Bob",   "age": 20},
]

sorted(users, key=lambda u: (u["name"], u["age"]))
````

This is the kind of thing that would be painful in many languages — but trivial in Python.
