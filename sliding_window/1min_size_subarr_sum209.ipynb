### 209. Minimum Size Subarray Sum (Medium)

Given an array of positive integers `nums` and a positive integer `target`, return the **minimal length** of a **subarray** whose sum is *greater than or equal to* `target`. If there is no such subarray, return `0` instead.

Example 1:
```shell
Input: target = 7, nums = [2,3,1,2,4,3]
Output: 2
```
Explanation: The subarray `[4,3]` has the minimal length under the problem constraint.

Example 2:
```shell
Input: target = 4, nums = [1,4,4]
Output: 1
```

Example 3:
```shell
Input: target = 11, nums = [1,1,1,1,1,1,1,1]
Output: 0
```

### Reasoning
What are the two conditions that control whether you **expand** or **shrink** the window in that problem?
> The algorithm always expands first (the `for` loop), then shrinks inside a `while` - it guarantees `r` always moves forward, no condition needed. The shrinking handles itself.

`for` + `while` version is the standard, cleaner template.

### Optimized Sliding Window — $O(n)$
**Core idea**:

Use **two pointers** (`left`, `right`) and a **running sum**.
- Expand `right` to increase the sum
- Shrink `left` to minimize the window once the `sum ≥ target`

**This is the canonical solution**.

````python
def min_adjacent_sliding_window(nums: list[int], target: int) -> int:
    # setup:
    # keep track of number of items: min_len = float('inf')
    # keep trach of window sum: tot = 0
    # set l pointer to 0
    # set r pointer - for loop - full scan

    # 1. exapnd window - tot += nums[r] - update min_len
    # 2. if tot covers t - shrink window - tot -= nums[l]
    # move l += 1
    # 3. return min_len if it's != float('inf') otherwise 0
    l = 0
    tot = 0
    min_len = float('inf')

    for r in range(len(nums)):
        tot += nums[r]  # expand window

        # shrink window while it still satisfies the target
        while tot >= target:
            min_len = min(min_len, r - l + 1)
            tot -= nums[l]
            l += 1

    return min_len if min_len != float('inf') else 0
````

````python
min_adjacent_sliding_window([2, 3, 1, 2, 4, 3], 7)
# func([1,4,4], 4)
# func([1,1,1,1,1,1,1,1], 11)
````

---

**Why this works**

Because all numbers are **positive**, the sum only increases when you move right, and only decreases when you move left.
This makes the sliding window valid and efficient.

**Time**:
- $O(n)$ — `right` moves `n` times total (the `for` loop)
- `left` moves at most `n` times total across the whole run (never resets).
> So total pointer movements `≤ 2n` → $O(n)$, not $O(n²)$ despite the nested loop.

**Space**: $O(1)$ — only `left`, `total`, `min_len`. No structure scales with input.

**When to use**
- Real interviews
- Production code
- Large datasets

### 🧠 How to Remember the Sliding Window Logic Forever
Think of it like **loading trucks**:
- Move `right` to add more warehouses until you have enough capacity
- Then move `left` to unload unnecessary warehouses and shrink the segment
- Keep track of the smallest valid segment

This mental model makes the algorithm intuitive.

### Short, interview‑ready explanation
“The brute‑force solution tries every possible adjacent segment and checks if its sum reaches the target, which is $O(n²)$. The optimized solution uses a sliding window: expand the right pointer to reach the target, then shrink the left pointer to minimize the window. Because each pointer moves only forward, the time complexity is $O(n)$.”

### Brute Force Solution — $O(n²)$

Core idea:
- Try every possible starting point, and from each, expand forward until the `sum ≥ target`.

This guarantees correctness but is **slow**.

````python
def min_subarray_len(nums, target):  # [2, 3, 1, 2, 4, 3], 7
    min_len = float('inf')

    for start in range(len(nums)):  # choose starting point
        total = 0
        for end in range(start, len(nums)):  # expand to the right
            total += nums[end]

            if total >= target:
                min_len = min(min_len, end - start + 1)
                break  # no need to expand further for this start

    return min_len if min_len != float('inf') else 0
````

````python
min_subarray_len([2, 3, 1, 2, 4, 3], 7)
````

- It checks all possible adjacent segments
- Stops early when a segment reaches the target
- But still worst‑case $O(n²)$

When to use
- Small inputs
- When learning the problem
- When verifying correctness of optimized solutions
