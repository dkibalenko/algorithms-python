### Sliding Window

Sliding Window works because **adjacent windows overlap** — you carry state forward instead of recomputing. The window is the mechanism; the overlap property is what makes it $O(1)$ per step.

> **The Core Insight**: Instead of recomputing a result from scratch for every subarray,
maintain a running result and update it incrementally as the window moves. The redundant work in brute force is that adjacent subarrays share almost all their elements — you recompute the same elements over and over. The sliding window eliminates that.

**The Mental Model**
```shell
Array:  [2, 4, 1, 7, 3, 5]
         ↑           ↑
       left         right     ← two pointers define the window

```
The window SLIDES right. On each step:
- Add the new element entering from the right
- Remove the old element leaving from the left
- Update the running result
```shell
Cost per step: O(1)
Cost total:    O(n)   ← each element added once, removed once
vs brute force O(n²)  ← recomputes the whole window every time
```

**Two Variants**
```shell
FIXED SIZE WINDOW              VARIABLE SIZE WINDOW
──────────────────────         ────────────────────────────
Window size k is given         Window shrinks/grows by condition
right - left == k always       right - left varies

right moves every step         right always moves forward
left = right - k + 1           left moves when condition violated

"max sum of k elements"        "smallest subarray with sum ≥ target"
"k-length substring with       "longest substring without repeating
 most vowels"                   characters"
```

**The Two-Line Core**
```py
# Every sliding window reduces to this:

for right in range(len(arr)):
    # 1. EXPAND — add arr[right] to window state

    while window_violates_condition:
        # 2. SHRINK — remove arr[left] from window state
        left += 1

    # 3. UPDATE answer from current valid window
```

**When to Reach for It**
> The signal is this question: *"Does the problem ask about a contiguous subarray or substring that satisfies some condition?"*
```shell
  → sum, product, average of subarray     → sliding window
  → longest/shortest subarray where X     → sliding window
  → substring with at most K distinct     → sliding window
  → fixed-size window maximum             → sliding window + deque
```

---

**Space Complexity**:
  - $O(1)$ if numeric (running sum)
  - $O(alphabet\space size)$ if tracking characters (set or dict) — but bounded by constant, so treated as $O(1)$

Sliding Window's enemy is **recomputing a subarray from scratch** every time you move. Look at the brute force for "find subarray with `sum ≥ target`":

```shell
[2, 3, 1, 2, 4, 3],  target = 7

Start at index 0:  2+3+1+2 = 8 ✓  (computed 4 elements)
Start at index 1:  3+1+2+4 = 10 ✓ (recomputed 4 elements — but 3,1,2 were already summed!)
Start at index 2:  1+2+4   = 7 ✓  (recomputed again from scratch)
```

> That overlap — recomputing elements you already saw — is the waste. $O(n²)$.

> **Sliding Window's insight:** when the window moves one step right, you don't recompute. You do **two $O(1)$** operations:

```shell
Add the new right element   → expand
Remove the old left element → shrink
```

```shell
[2, 3, 1, 2, 4, 3]
 L           R      sum=8, too big → remove left (2), L moves right
    L        R      sum=6, too small → add right (3), R moves right
    L           R   sum=9 ...
```

One pass. $O(n)$.

---

#### Sliding Window — Two Patterns


> **FIXED window**: size determined by `s1`, slides by 1 each step → build initial window, then add right / remove left

> **VARIABLE window**: size determined by a **condition** (sum, uniqueness) → expand always, shrink while condition violated

**Both**:
- $O(n)$ time — each element added **once**, removed **once**
- Each step is $O(1)$ — **no recomputation** from scratch
- Dict comparison bounded by alphabet size → $O(1)$

**When to recognize it**:
- "substring/subarray satisfying a condition" or,
- "does X contain Y as a permutation/anagram"

---

**Two Pointers:**

- Flavor 1 (opposite ends): Two Sum II, Valid Palindrome, 3Sum
- Flavor 2 (same direction): Move Zeros, Remove Duplicates, Merge Sorted Array
- You understand the directional decision logic, amortized O(n), and when sorting is a prerequisite

**Sliding Window:**

- Variable window: Min Size Subarray Sum, Longest Substring Without Repeats
- Fixed window: Permutation in String
- You understand the expand/shrink mechanics, why it's O(n) despite nested loops, and the frequency dict pattern

That's the full range of both patterns.

### Two Pointers & Sliding Window Takeaways
They solve primarily $O(n^2)$ quadratic (cubic, etc.) problems, letting us to avoid nested loops that make n operations for each item on the input n size, by utilizing two pointers at the ends or same direction (+3 pointers with the fixed one), and sliding window pattern either with hash table data structure like dict or set, fixed or variable window making the same $O(n^2)$ problem more efficient with $O(n)$ + extra space when needed.

Two additions worth filing before you close:

**1. What enables the optimization is structure, not just pointers.**

Two Pointers works because **sorted order** (or a maintained invariant) lets you make a directional decision — *"too big, move right; too small, move left."* Without that structure you're back to nested loops. The pointers are the mechanism; the structure is the prerequisite.

Sliding Window works because **adjacent windows overlap** — you carry state forward instead of recomputing. The window is the mechanism; the overlap property is what makes it $O(1)$ per step.

**2. Space is O(1) for Two Pointers, not always for Sliding Window.**
```shell
Two Pointers:     always O(1) space — just index variables
Sliding Window:   O(1) if numeric (running sum)
                  O(alphabet size) if tracking characters (set or dict)
                  — but bounded by constant, so treated as O(1)
```
