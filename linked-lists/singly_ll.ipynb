### Singly Linked List

```python
class Node:
    def __init__(self, value) -> None:
        self.value = value
        self.next = None

    def __str__(self) -> str:
        return (
            "({class_name} value: {value}, next: {next})".format(
                class_name=__class__.__name__,
                value=self.value,
                next=self.next
            )
        )
# []

class LinkedList:
    def __init__(self, value) -> None:
        self.head = Node(value)
        self.tail = self.head
        self.length = 1

    def __str__(self) -> str:
        return (
            "({class_name}, Head: {head}, Tail: {tail}, Length: {length})".format(
                class_name=__class__.__name__,
                head=self.head,
                tail=self.tail,
                length=self.length
            )
        )

    def append_item(self, value) -> None:
        new_node = Node(value)  # Node(value=value, next=None)

        self.tail.next = new_node  # connect tail to new node
        self.tail = new_node  # move tail to the last new node
        self.length += 1
```
