### Two Pointers

Two Pointers works because **sorted order** (or a maintained invariant) lets you make a directional decision — *"too big, move right; too small, move left."* Without that structure you're back to nested loops. The pointers are the mechanism; the structure is the prerequisite.

**Space complexity**: always $O(1)$ space — just index variables

**The Problem Two Pointers Exists to Avoid**

Every Two Pointer solution is replacing this:
```py
# Find a pair that sums to target
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):   # ← the crime
        if nums[i] + nums[j] == target:
            return [i, j]
```

What's wrong with this? For an array of 1,000 elements, you check ~500,000 pairs. For 10,000 elements — ~50,000,000 pairs. That's $O(n²)$.

The question is: **can we skip pairs we'll never need**?

---

**The Core Insight — Why It Works on Sorted Arrays**

Take a sorted array. You're looking for a pair that sums to `9`:

```shell
[ 1,  2,  4,  6,  8,  11 ]
  L                    R
```

`1 + 11 = 12`. Too big.

Here's the key question — **which pointer should move, and why?**

Think about it before reading on. The array is sorted. You need a smaller sum. Can you get a smaller sum by moving `L` right? Or by moving `R` left?

Moving `L` right makes the sum _larger or equal_. Moving `R` left makes it _smaller_. So move `R`:

```shell
[ 1,  2,  4,  6,  8,  11 ]
  L               R
```

`1 + 8 = 9`. Found it.

> **That decision — "too big → move R left, too small → move L right" — is the entire pattern.** Each move eliminates every pair containing the discarded element. You skip pairs without checking them.

---

**The Invariant (This Is What You Were Missing)**

> Two Pointers works when you can answer: _"given the current result, which direction gets me closer to the target?"_

That requires **structure** in the data — usually sorted order, or an invariant you maintain yourself. Without structure, you can't make that directional decision, and you're back to nested loops.

This is why your complexity instinct was right but incomplete:

```shell
Brute force (nested loops):  O(n²) time — check every pair
Two pointers (sorted input):  O(n)  time — each pointer moves at most n steps total
                              O(1)  space — just two index variables, no extra storage
```

### The Four Patterns You'll See in Interviews

PATTERN 1 — Opposite ends (single array, converge)
```shell
  [ 1, 2, 4, 6, 8, 11 ]
    L →           ← R
```
- Pointers converge. Stop when they meet.
- Used for: **Two Sum II, Valid Palindrome**

PATTERN 2 — Same direction (slow/fast, single array)
```shell
  [ 0, 1, 0, 3, 0, 2 ]
    S
    F →
```
- Fast reads, slow writes. Stop when fast reaches end.
- Used for: **Remove Duplicates, Move Zeros**

PATTERN 3 — Fixed pointer + inner two-pointer (extension of Pattern 1, same array)
```shell
  [ -4, -1, 0, 1, 2, 4 ]
     i   L →       ← R
```
- Outer loop fixes one element, inner two-pointer (Pattern 1) finds the pair against it.
- Used for: **3Sum**

PATTERN 4 — Multi-pointer across two arrays (backward merge)
```shell
  nums1: [ 1, 2, 3, _, _, _ ]  nums2: [ 2, 5, 6 ]
                x →                    y →
                  (both read backward, write pointer i also backward into nums1)
```
- Two read pointers (one per array) plus a write pointer, all moving backward. Stop when each sub-array is exhausted.
- Used for: **Merge Sorted Array**

---

**Two Pointer pattern summary — you're solid on:**

|Flavor|Problem|Direction|Stop when|
|---|---|---|---|
|Opposite ends|Two Sum II, Valid Palindrome|converge|`left < right`|
|Same direction|Move Zeros, Remove Duplicates|both right|`fast == len`|
|Fixed + inner two-pointer|3Sum|outer forward, inner converge|inner `left < right`|
|Multi-pointer (two arrays)|Merge Sorted Array|all backward|each sub-array exhausted|

### What Redundant Work Each Pattern Actually Avoids

The intro at the top of this notebook only tells Pattern 1's story — nested-loop pair-checking, eliminated by sorted order. The other patterns don't avoid that at all, and Pattern 2 doesn't even require sorted input. Each has its own redundant-work story:

**Pattern 1 — Opposite ends** (Two Sum II, Valid Palindrome)
```shell
Naive:       nested loop checks every pair                → O(n²) time
Two Pointer: sorted order lets you discard a whole side    → O(n)   time, O(1) space
```
> Avoids: re-checking pairs that sortedness already rules out.

**Pattern 2 — Same direction** (Move Zeros, Remove Duplicates)
```shell
Naive:  build a separate result array                 → O(n) space
   or:  repeatedly pop/delete from the middle of nums  → O(n) per removal, O(n²) worst case
Two Pointer: write in place in one pass                → O(n) time, O(1) space
```
> Avoids: a second array, **and** the shifting cost of removing/inserting mid-array. `slow` never scans back — it just marks "the next correct write position," so nothing needs re-checking as `fast` advances. This pattern needs **no sorted input at all** — it works on any array, because its invariant isn't about value order, it's "everything before `slow` is already correctly placed."

**Pattern 3 — Fixed + inner two-pointer** (3Sum)
```shell
Naive:       three nested loops check every triplet        → O(n³) time
Two Pointer: fix one element, sorted inner two-pointer      → O(n²) time, O(1) extra space (excl. sort)
```
> Avoids: the third loop entirely — for each fixed element, Pattern 1 finds the remaining pair in one linear pass instead of another nested scan.

**Pattern 4 — Multi-pointer across two arrays** (Merge Sorted Array)
```shell
Naive:     copy both arrays out, sort the combined result   → O((m+n) log(m+n)) time, O(m+n) space
Multi-Pointer: write from the back, no sort needed           → O(m+n) time, O(1) space
```
> Avoids: re-sorting. Since **both inputs are already individually sorted**, you only ever need to compare the two current frontier elements — writing the largest remaining value into place, back to front, guarantees correctness without touching an already-placed element. This pattern needs sorted input, same as Pattern 1 — but it exploits sortedness to avoid a *sort*, not to avoid *pair-checking*.
