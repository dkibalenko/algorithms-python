### 26. Remove Duplicates from Sorted Array
Given an integer array `nums` sorted in **non-decreasing order**, remove the duplicates **in-place** such that each unique element appears only **once**. The **relative order** of the elements should be kept the **same**.

Consider the number of _unique elements_ in `nums` to be `k**​​​​​​​**`​​​​​​​. After removing duplicates, return the number of unique elements `k`.

The first `k` elements of `nums` should contain the unique numbers in **sorted order**. The remaining elements beyond index `k - 1` can be ignored.

**Example 1:**
```shell
Input: nums = [1,1,2]
Output: 2, nums = [1,2,_]
```
**Explanation:** Your function should return k = 2, with the first two elements of nums being 1 and 2 respectively.

It does not matter what you leave beyond the returned k (hence they are underscores).

**Example 2:**
```shell
Input: nums = [0,0,1,1,1,2,2,3,3,4]
Output: 5, nums = [0,1,2,3,4,_,_,_,_,_]
```
**Explanation:** Your function should return k = 5, with the first five elements of nums being 0, 1, 2, 3, and 4 respectively.

It does not matter what you leave beyond the returned k (hence they are underscores).

**Constraints:**

- `1 <= nums.length <= 3 * 10^4`
- `-100 <= nums[i] <= 100`
- `nums` is sorted in **non-decreasing** order.

### Reasoning
#### Step 1 - Gather Requirements

- In the given integer array, sorted ASC, I need to remove all duplicates in-place, such only **unique elements remaining**. I need to maintain the relative order of those unique items. I need to return the **number of unique items** `k`. I need to do it in-place, so I think I need to move all unique items to the beginning of the array, leaving and ignoring duplicates behind.
- **Input**: integer array, sorted ASC, has duplicates, negatives, zeros `-100 - 100`
- **Output**: integer - length of the unique items of the array
- **Empty/null input**: No

#### Step 2 - Examples + Edge cases
1. `[1]` -> `[1]` -> `1` output
2. `[2, 2, 2]` -> `[2]` -> `1` output
3. `[1, 2, 3, 4]` -> `[1, 2, 3, 4]` -> 4 output
4. `[0, 0, 1, 1, 1, 2, 2, 3, 3, 4]` → `[0, 1, 2, 3, 4]` -> 5 output

#### Step 3 - brute force
Use a set data structure to deduplicate the input array + copy it back to input.

````python
def func(nums: list[int]) -> int:
    # brute force - use a set for dropping duplicates
    # setup

    # 1. unpack the input into the set data structure with list compreh.
    # 2. unpack the set back into the input list object and sort it
    # 2. return length of the input list of unique items

    temp = {item for item in nums}
    nums[:] = sorted(temp)  # slice assignment to preserve the input object

    return len(nums)
````

````python
# func([1])
# func([2, 2, 2])
# func([1, 2, 3, 4])
# func([0, 0, 1, 1, 1, 2, 2, 3, 3, 4])
func([-1, -1, 0, 0, 2, 2, 5, 5])
````

> **Brute force trades a sort you don't actually need**, since the input is _already sorted_ — duplicates are guaranteed to sit adjacent to each other, so a single forward pass can detect and skip them without ever re-sorting.

- **Time**: $O(n\space log\space n)$ — set-build is $O(n)$, but sorted() on up to `n` unique elements dominates at $O(n\space log\space n)$
- **Space**: $O(n)$ — worst case (no duplicates) the set and sorted list both hold `n` elements

vs. optimized two-pointer: $O(n)$ time, $O(1)$ space

#### Step 4 - Select Algorithm + state complexity

**Two pointers - Same direction** (slow/fast). **One pass, sorted array** - reading elements (`fast` pointer) & writing in-place (`slow` pointer). $O(n)$ time and constant space $O(1)$, since the write is done **in-place** of the input array.

Set two pointers - `slow` at 0 and `fast` at 1. The `fast` pointer is pushed forward to **read** by for loop, to find the distinct element from `slow`. Once `fast` has found it, `slow` knows that another unique element is comming, so it is moved forwar for writing position. `fast` is written to `slow`. No redundant extra counter is used - `slow + 1` is self-evident and counts the total number of unique items found.

- `slow + 1` count. This is self-descriptive pointer, which is moved (counts) every time the unique item is found by fast. That is why it could be directly be used to express the number of unique items in the array.

Exactly right. `slow` counts how many times you've advanced to write a unique element — so its final value is the **index of the last unique element written**, and `slow + 1` converts that 0-based index to a count.
> The pointer does double duty: it writes and it counts simultaneously.

**Step 7 - write code**

````python
def remove_duplicates(nums: list[int]) -> int:
    # two pinters - same dir
    # setup
    # set slow pointer to 0
    # set fast pointer to 1 - for loop [1, len(nums)]

    # 1. check if slow != fast - unique item is found ->
    #   a) move slow by 1
    #   b) write fast at slow
    # 2. return slow + 1 - slow is at index of the last unique item

    slow = 0

    for fast in range(1, len(nums)):
        if nums[slow] != nums[fast]:
            slow += 1
            nums[slow] = nums[fast]

    return slow + 1
````

````python
# remove_duplicates([1, 1, 2])  # 2
remove_duplicates([0, 0, 1, 1, 1, 2, 2, 3, 3, 4])  # 5
````

#### Step 9 - Complexity analysis

One pass with slow and fast pointers - read unique items by fast & write them with slow - $O(n)$ time. No extra space is used, the job is done in-place with constant $O(1)$ space.
