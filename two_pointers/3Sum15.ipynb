### 15. 3Sum
Given an integer array `nums`, return all triplets `[nums[i], nums[j], nums[k]]` such that they sum to zero and no duplicate triplets appear in the result.

(Given an integer array `nums`, return all the triplets `[nums[i], nums[j], nums[k]]` such that `i != j`, `i != k`, and `j != k`, and `nums[i] + nums[j] + nums[k] == 0`.

Notice that the solution set must not contain duplicate triplets.)

Example 1:
```shell
Input: nums = [-1,0,1,2,-1,-4]
Output: [[-1,-1,2],[-1,0,1]]
Explanation: 
nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0.
nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0.
nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0.
The distinct triplets are [-1,0,1] and [-1,-1,2].
```
Notice that the order of the output and the order of the triplets does not matter.

Example 2:
```shell
Input: nums = [0,1,1]
Output: []
Explanation: The only possible triplet does not sum up to 0.
```
Example 3:
```shell
Input: nums = [0,0,0]
Output: [[0,0,0]]
Explanation: The only possible triplet sums up to 0.
```

Two questions:
1. What's the first thing you'd do to the array, and why does that unlock the Two Pointer approach?
2. How does sorting help you skip duplicate triplets without extra logic?

- The array must be sorted before we use it with two pointers with Flavor 1.

> **Sort first, then two pointers**.

Now the duplicate insight — let me show you with a concrete example.

After sorting `[-4, -1, -1, 0, 1, 2]`, your outer loop fixes `i`:
```shell
i=0: nums[i] = -4  → two pointer on the rest
i=1: nums[i] = -1  → two pointer on the rest
i=2: nums[i] = -1  → SAME VALUE as i=1, already checked all pairs for -1
```
If you don't skip `i=2`, you produce duplicate triplets. But how do you know to skip it?
- Because after sorting, **duplicates are always adjacent**.
One check handles it:
```py
if i > 0 and nums[i] == nums[i - 1]:
    continue
```

The three cases are exactly right. Now let me make the duplicate-on-match concrete.

After sorting, say you find a valid triplet and append it. You then move `left += 1` and `right -= 1`. But look at this input:
```shell
[ -2,  0,  0,  2,  2 ]
   i   L        R
```
`-2 + 0 + 2 = 0` → match, append `[-2, 0, 2]`. Now you move both inward:

```shell
[ -2,  0,  0,  2,  2 ]
   i       L   R
```
`-2 + 0 + 2 = 0` → match again. But it's the same triplet.

The values at the new positions are identical to what you just used. So the question is: *after moving both pointers inward, **what single check** tells you "I've landed on a duplicate, keep moving"?*

> Hint: you already wrote the same logic for the outer `i` pointer. Think about what happened just before: you already moved left += 1. So you're now at the new position. The duplicate would be behind you, not ahead.

The check is:

```python
while left < right and nums[left] == nums[left - 1]:
    left += 1
```

> *"Did I just land on the same value I used in the last match? Keep moving."*

Same logic for right — looking right (forward), since you moved inward:

```python
while left < right and nums[right] == nums[right + 1]:
    right -= 1
```

Now you have everything. Write the complete `while left < right` block — all three cases plus duplicate skipping on match.

> **CORE IDEA**: for each fixed pointer's value (`i`), check all possible pairs with another 2 pointers.

````python
def three_sum(nums: list[int]) -> list[list[int]]:
    # [-1, 0, 1, 2, -1, -4] -> [-4, -1, -1, 0, 1, 2]
    # SETUP:
    # sort the array
    # set res list
    # scan the array: len(arr) - 2 -> fixed pointer i
    # check for fixed pointer value duplicates - skip if needed
    # set l pointer to i + 1, and r pointer to len(arr) - 1

    # FOR EACH FIXED POINTER value (i), check all pairs with 2 pointers:
    # 1. compute the sum
    # 2. handle 3 cases:
    # case 1: match - append to res; move L, R inward; check L, R for duplicates - move if needed
    # case 2: sum < 0 - move L forward
    # case 3: sum > 0 - move R backward
    # return res
    nums.sort()
    res = []

    for i in range(len(nums) - 2):
        # skip duplicates for fixed 'i' pointer
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        l, r = i + 1, len(nums) - 1

        while l < r:
            total = nums[i] + nums[l] + nums[r]

            # handle 3 cases:
            if total == 0:
                res.append([nums[i], nums[l], nums[r]])

                # move both inward
                l += 1
                r -= 1

                # skip duplicates for l, r pointers
                while l < r and nums[l] == nums[l - 1]:
                    l += 1
                while l < r and nums[r] == nums[r + 1]:
                    r -= 1
            elif total < 0:
                l += 1
            else:
                r -= 1
    return res
````

````python
three_sum([-1, 0, 1 , 2, -1, -4])
````

**Mental Model**

> 3Sum = Sort + (for each element: Two Sum II on the remainder)

1. Sort the array                              → enables two pointers + duplicate skipping
2. Fix `i`, set `left = i+1`, `right` = end    → reduces to a 2-pointer subproblem
3. Three cases:
    - `sum < 0`  → `left  += 1`  (need bigger)
    - `sum > 0`  → `right -= 1`  (need smaller)
    - `sum == 0` → record, move both inward, skip adjacent duplicates
4. Skip duplicates on `i`:
    - `nums[i] == nums[i-1]`                    → continue
5. Skip duplicates on match:
    - `nums[left] == nums[left-1]` after moving
    - `nums[right] == nums[right+1]` after moving

> Pattern name to file: **Two Pointers — Flavor 1 with outer loop (reduces N-sum to (N-1)-sum)**.

---

**Time: $O(n²)$**

- **Sort**:          $O(n\space log\space n)$
- **Outer loop**:    $O(n)$
- **Inner 2ptr**:    $O(n)$ per outer iteration

**Total**: $O(n\space log\space n) + O(n²) = O(n²)$   ← **$n²$ dominates**

**Space: two answers depending on what you count**

- **Auxiliary space**:  $O(log\space n)$ — just the **sort's call stack**, no extra structures
- **Output space**:     $O(n²)$    — worst case, every triplet is valid

In interviews, when they say "space complexity" they usually mean auxiliary (excluding output). So the clean answer is **$O(log\space n)$ auxiliary, $O(n²)$ output**.
