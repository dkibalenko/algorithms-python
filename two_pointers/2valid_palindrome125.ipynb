### 125. Valid Palindrome
A phrase is a **palindrome** if, after converting all uppercase letters into lowercase letters and removing all non-alphanumeric characters, it reads the same forward and backward. Alphanumeric characters include letters and numbers.

Given a string `s`, return `true` _if it is a **palindrome**, or_ `false` _otherwise_.

Example 1:
```shell
Input: s = "A man, a plan, a canal: Panama"
Output: true
Explanation: "amanaplanacanalpanama" is a palindrome.
```
Example 2:
```shell
Input: s = "race a car"
Output: false
Explanation: "raceacar" is not a palindrome.
```
Example 3:
```shell
Input: s = " "
Output: true
Explanation: s is an empty string "" after removing non-alphanumeric characters.
Since an empty string reads the same forward and backward, it is a palindrome.
```

**Constraints**:
```shell
1 <= s.length <= 2 * 10^5
s consists only of printable ASCII characters.
```

- **Input**: string up to 100000 chars, can contain non-alphanumeric chars, can be empty string
- **Output**: boolean

**Step 3 - brute force**

def func(s: str) -> bool:
    clean_s = ""

    for c in s:
        if c.isalnum():
            clean_s += c.lower()

    l, r = 0, len(clean_s) - 1

It uses extra memory to build and store the new clean string by filtering non-alphanumeric chars. So space is $O(n)$. Time is $O(n + k)$ where `k` is the new cleaned string.

**1. `.isalnum()` & reverse string with slicing. Time $O(n)$, Space $O(n)$**

````python
def is_palindrome(string: str) -> bool:
    result_string = ""  # build a new string

    for char in string:
        if char.isalnum():  # allow only letters & numbers
            result_string += char.lower()

    return result_string == result_string[::-1]
````

**Step 4 - select algorithm + state complexity**

Two pointers - opposite ends. After turning the input string into lowercase, set two pointers at the ends of it. Then 2 checks:
- filter. if chars not alphanumeric - skip them
- if chars not the same - return `False`, otherwise move pointers and return `True` at the end meaning all pass.

> **Key insight**: One $O(n)$ pass with string's `.lower()` method to convert all letters into lowercase & another one $O(n)$ pass with two pointers to visit all chars & compare againsts each other, skipping non-alphanumerics.

**Step 7 - write code**:

**2. Two Pointers - Flavor 1. Time $O(n)$, Space $O(1)$**

````python
def is_palind(string: str) -> bool:
    # two pointers - oppos ends
    # setup
    # lowercase input
    # set 2 pointers

    # CONDITION: l < r do job

    # 1. check l, r - skip non-alphanums
    # 2. if l != r -> False
    # 3. move pointers
    # 4. return True - happy path

    string = string.lower()              # O(n)
    left, right = 0, len(string) - 1

    while left < right:                  # O(n)
        # 1. skip non-alphanumerics
        if not string[left].isalnum():
            left += 1
            continue
        if not string[right].isalnum():
            right -= 1
            continue

        # 2. stop early when not equal
        if string[left] != string[right]:
            return False
        # 3. move pointers
        left += 1
        right -= 1

    # 4. happy path - valid palindrome
    return True
````

**Step 9 - complexity analysis**

Time complexity is $O(n + n)$ because of two $O(n)$ passes one for string's `lower()` method to turn the input string into lowercase & another one pass to scan the input for palindrom check. Space complexity is $O(1)$ constant, no extra space is used.

````python
def is_alpha_num(char: str) -> bool:
    return (
        ord('A') <= ord(char) <= ord('Z')
        or ord('a') <= ord(char) <= ord('z')
        or ord('0') <= ord(char) <= ord('9')
    )


def is_palindrome2(string: str) -> bool:
    left, right = 0, len(string) - 1  # two pointers at both ends

    while left < right:
        # move pointers, to make sure both chars are alphanumeric
        while left < right and not is_alpha_num(string[left]):  # move left if not alphanumberic
            left += 1
        while left < right and not is_alpha_num(string[right]):  # move right if not alphanumeric
            right -= 1

        # compare alphanumeric chars at both pointers
        if string[left].lower() != string[right].lower():
             return False  # when chars don't match - it's not a palindrome

        # move pointers to each other
        left, right = left + 1, right - 1

    # loop with test is done, means it's a palindrome
    return True
````
