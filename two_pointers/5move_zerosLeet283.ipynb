### 283. Move Zeroes

[Link](https://leetcode.com/problems/move-zeroes/description/?envType=problem-list-v2&envId=oipsks9i)

Given an integer array `nums`, move all `0`'s to the end of it while maintaining the relative order of the non-zero elements.

Note that you must do this in-place without making a copy of the array.

Example 1:
```shell
Input: nums = [0,1,0,3,12]
Output: [1,3,12,0,0]
```
Example 2:
```shell
Input: nums = [0]
Output: [0]
```

**Constraints**:
```shell
1 <= nums.length <= 10^4
-2^31 <= nums[i] <= 2^31 - 1
``` 

**Follow up**: Could you minimize the total number of operations done?

### Reasoning
#### Step 1 - Restate the problem yourself

You're not finding a pair. No target sum involved.
- I need to move all zeros in the given integer array to the end of the array, while maintaning the relative order of all non-zero numbers.
- **Input**: an integer array, not ordered, has duplicates and negatives(likely)
- **Output**: the same integer array, the job is done in-place.

#### Step 2 - Edge case + example
1. **Case A**: `[0, 1, 0, 3, 12]` returns `[1, 3, 12, 0, 0]`
2. **Case B**: `[0]` returns `[0]`
3. **Case C**: Empty input -> `[]`
4. **Case D**: `[2, 2, 2, 2]` returns `[2, 2, 2, 2]`
5. **Case E**: `[0, -2, 4, 0, -5, 9]` returns `[-2, 4, -5, 9, 0, 0]`

#### Step 3 - Brute force

What's the naive approach, what's its time complexity, and what's its space complexity?
- For naive approach, although the problem requires 'in-place' solution, I would try to have another array to keep all zeros from the input one. It doesn't use some extra space making the solution's space complexity $O(n)$ since it pops zeros and adds them to this array keeping space capacity intact. The only problem it makes is `.pop(index) ` method for array. It degrades $O(n)$ time to $O(n * k)$ where k is the number of zeros in the input. Potentially, if all zeros in the array it falls to $O(n^2)$, since for each element it would pop by index (which is $O(n)$ itself because of recomputing the input array's indices every time it pops not from the end). And I'm not sure about `.extend()` method whether it $O(n)$ or $O(1)$.

Good analysis. `.extend()` is $O(k)$ where `k` is the **number of zeros** — it copies each element once. Not $O(1)$, but it only runs once, so it doesn't change the overall complexity.

Your brute force is correct and your complexity reasoning is right:

```shell
Time:  O(n²) worst case — pop(index) shifts all elements right of it
Space: O(n)  — zeros array grows with number of zeros in input
```

````python
def func(nums: list[int]) -> list[int]:
    zeros = []
    l = 0

    while l < len(nums):
        if nums[l] == 0:
            zeros.append(nums.pop(l))
        else:
            l += 1

    nums.extend(zeros)

    return nums
````

````python
# func([0, 1, 0, 3, 12])  # [1, 3, 12, 0, 0]
# func([2, 2, 2, 2])  # [2, 2, 2, 2]
func([1, 0, -2, 4, 0, -5, 9])  # [1, -2, 4, -5, 9, 0, 0]
# func([0])  # [0]
# func([])  # []
````

### Step 4 — Select Algorithm
![Visual](../images/283MoveZeros.png)

The problem says in-place, which means the target is $O(1)$ space.

**Pattern**: `slow` **carries** foward 0 or **the last 0 of the zeros group** in order for `fast` pointer to make a **swap** when **non-zero** item is found.

**State the key insight first**: what invariant do the two pointers maintain, and why does that eliminate the need for a separate zeros array?
- The **invariant** the two pointers maintain is that, `slow` always points to the **left-most zero** to be moved to the right, and `fast` pinter **reads** forward to find **non-zero** elements to **swap** with the `slow`.

The two pointers approach - Pattern 2 (fast - read, slow write) mutates the input array in-place, maintaining it as invariant. It totally eliminates the need for a separate zero arary, since it makes ONE FULL PASS over unsorted array.

> Non-zero items are **read by the `fast` pointer**, and are **swapped** with the `left` one when non-zero if found.

Slow pointer moves behind the fast one, always pointing to the left-most zero left behind the `fast` pointer and prepare them for the next swap. So, no serparate array is needed, one pass with swaps makes this approach constant space $O(1)$.

> Everything at index < `s` is a correctly placed non-zero. `s` always points to the next position where a non-zero should be written. Whatever sits at `s` is either a zero or hasn't been processed yet.

Total time complexity is $O(n)$ linear - one full pass is made with read and write ops, space $O(1)$ - the input array is mutated in-place.

```shell
Move Zeros — Two Pointers Flavor 2
Invariant:  everything before slow is a correctly placed non-zero
Brute force: O(n²) time, O(n) space — pop(index) shifts elements
Optimized:   O(n)  time, O(1) space — single pass, zeros left behind
Key insight: non-zeros written forward → zeros naturally end up at the back
             no separate array needed, no shifting
```

#### Step 7 - Write code

````python
def move_zeros(nums: list[int]):
    s = 0

    for f in range(len(nums)):
        if nums[f]:
            nums[s], nums[f] = nums[f], nums[s]
            s += 1
    return nums
````

````python
# move_zeros([0, 1, 0, 3, 12])  # [1, 3, 12, 0, 0]
# move_zeros([2, 2, 2, 2])  # [2, 2, 2, 2]
move_zeros([1, 0, -2, 4, 0, -5, 9])  # [1, -2, 4, -5, 9, 0, 0]
# move_zeros([0])  # [0]
# move_zeros([])  # []
````

````python
def movezeros(nums):
    slow = 0
    fast = 0

    while fast < len(nums):
        if nums[fast]:
            nums[slow], nums[fast] = nums[fast], nums[slow]
            slow += 1

        fast += 1
    return nums
````
