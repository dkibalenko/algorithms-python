### 167. Two Sum II - Input Array Is Sorted
[Link](https://leetcode.com/problems/two-sum-ii-input-array-is-sorted/description/)

Given a **1-indexed** array of integers `numbers` that is already **_sorted in non-decreasing order_**, find two numbers such that they add up to a specific `target` number. Let these two numbers be `numbers[index1]` and `numbers[index2]` where `1 <= index1 < index2 <= numbers.length`.

Return _the indices of the two numbers_ `index1` _and_ `index2`_, **each incremented by one,** as an integer array_ `[index1, index2]` _of length 2._

The tests are generated such that there is **exactly one solution**. You **may not** use the same element twice.

Your solution must use only constant extra space.

Example 1:
```shell
Input: numbers = [2,7,11,15], target = 9
Output: [1,2]
Explanation: The sum of 2 and 7 is 9. Therefore, index1 = 1, index2 = 2. We return [1, 2].
```
Example 2:
```shell
Input: numbers = [2,3,4], target = 6
Output: [1,3]
Explanation: The sum of 2 and 4 is 6. Therefore index1 = 1, index2 = 3. We return [1, 3].
```
Example 3:
```shell
Input: numbers = [-1,0], target = -1
Output: [1,2]
Explanation: The sum of -1 and 0 is -1. Therefore index1 = 1, index2 = 2. We return [1, 2].
```

### Reasoning
**Step 1** - restate the problem in your own words.
- In the given integer array, which is **1-indexed**, I need to find 2 items that add up to the given target integer number. The array is sorted in ascending order. I need to return the indices of these two numbers incremented by 1 in the new integer array of length 2. The two valid items must be exactly distinct items. There might be exactly 1 solution. The solution must use constant space.
- **Input**: integer array, sorted ASC, negatives - Yes, duplicates - No.
- **Output**: integer array of length 2 or empty list.
- Input is not null or empty.
- Output order matters.

**Step 2** - edge cases + examples
- Empty input -> `[]`
- Negatives -> valid input, hadles as needed

**Step 3** - brute force
```py
def twoSum(numbers, target):
    for i in range(len(numbers) - 1):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                return [i + 1, j + 1]

    return []
```
- The brute force with nested-loop does n operations for each element, making it $O(n^2)$ time and $O(1)$ constant space. The redundant work is the inner loop **searching the array for a value**.

**Step 4** - select algotithm + state complexity
- The nested-loop approach could be solved with two pointers - Pattern 1 - opposite ends (sorted or palindrome-style). The inner loop is eliminated by **ONE PASS with two pointers at the ends of the array that converge**. This results in liner $O(n)$ total time and constant $O(1)$ space.

> Two pointers at the ends make it $O(n)$ total time, eliminating the inner for loop.

1. What does your code do when `nums[l] + nums[r]` is **too small**? Which pointer moves and which direction?
2. What does it do when the sum is **too big**?
- when the `nums[l] + nums[r]` is too small for target my left pointer is moved right - we increase the sum, and when the sum is too big the right pointer is moved to the left - decremented.

Exactly right.

> That **directional logic** IS the algorithm — everything else is just bookkeeping.

```shell
  [ 1, 2, 4, 6, 8, 11 ]
    L →           ← R
```

**Step 5, 6** - signature + psudocode or skeleton
```py
def two_sum(nums: list[int], t: int) -> list[int]:
    # setup
    # set left pointer to 0: l = 0
    # set right pointer to the end: r = len(nums) - 1
    # CONDITION: run until pointers converge

    # 1. compute sum of l, r items: nums[l] + nums[r]
    # 2. hadnle 3 cases:
    # a) sum = t -> return [l + 1, r + 1]
    # b) sum < t -> move left pointer: l += 1
    # c) sum > t -> move right pointer: r -= 1
    # 3. return [] - no valid pair is found
```

**Step 7** - write code

````python
# Two pointers
def two_sum(numbers: list[int], target: int) -> list[int]:
    l, r = 0, len(numbers) - 1

    while l < r:
        curr_sum = numbers[l] + numbers[r]
        if curr_sum == target:
            return [l + 1, r + 1]
        elif curr_sum < target:
            l += 1
        else:
            r -= 1

    return []
````

````python
# two_sum([2, 7, 11, 15], 9)  # [1,2]
# two_sum([3, 3, 3, 3], 6)  # [1, 4]
two_sum([-1, 0], -1)  # [1, 2]
````

**Step 9** - complexity analysis
- The time complexity of this algorithm is $O(n)$ - one pass with two pointers at the ends, checking the current sum against the give target. Space is constant $O(1)$ since no extra space is needed to store data, one vairable for current sum and returning 2-items list, that is it.

> **The pattern**: Two pointers - opposite ends (sorted & palindrome style). One pass with moving the pointers until they converge. Compute the current sum of two values and check against the target.
