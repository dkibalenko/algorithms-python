### 287. Find the Duplicate Number
Given an array of integers `nums` containing `n + 1` integers where each integer is in the range `[1, n]` inclusive.

There is only **one repeated number** in `nums`, return _this repeated number_.

You must solve the problem **without** modifying the array `nums` and using only **constant extra space**.

 

**Example 1**:
```shell
Input: nums = [1,3,4,2,2]
Output: 2
```
**Example 2**:
```shell
Input: nums = [3,1,3,4,2]
Output: 3
```
**Example 3**:
```shell
Input: nums = [3,3,3,3,3]
Output: 3
``` 

**Constraints**:
```shell
1 <= n <= 10^5
nums.length == n + 1
1 <= nums[i] <= n
```
All the integers in nums appear only once except for precisely one integer which appears two or more times.
 

**Follow up**:

How can we prove that at least one duplicate number must exist in `nums`?
Can you solve the problem in _linear runtime complexity_?

### Floyd's cycle detection

#### Step 1 — Gather Requirements
Given an integer array of `n + 1` within the range of elements from `1` to `n` inclusive. There is only one repeated number that appears twice or more times. I need to find & return this repeated element. The problem must be solved in-place without modifying the array and using extra space.

- **Sorted?** No — arbitrary order, not sorted at all.
- **Negatives/zeros?** No — the constraint `1 <= nums[i] <= n` guarantees every value is a positive integer ≥ 1.
- **Duplicates?** Yes, guaranteed — and per the constraints, exactly **one distinct value** is the duplicate, though (per your Example 3) it can occur more than twice.
- **Return type?** A single integer — the duplicate _value_ itself, not its index.

**The pigeonhole principle**

The array holds `n + 1` _items_ (slots in the array), but each item's _value_ can only be one of `n` possible values (`1` through `n`). If you have `n + 1` items and only `n` distinct values available to assign them, what does that force, by simple counting — regardless of what the actual values turn out to be? (This is the **pigeonhole principle** by name — `n+1` pigeons, `n` holes.)

That forces to **repeat** one of the elements **at least once**:
```shell
length = 5 + 1
values range = [1, 5]
arr = [1, 2, 3, 4, 5, 1] -> at least once item must repeat
```

> Given `n+1` slots and only `n` possible values, **at least one** value must repeat somewhere in the array.

#### Step 2 - Examples + Edge Cases
1. **Case**: `[1, 3, 4, 2, 2]` has length of 5 & since the first 4 elements are distinct, unsorted though, at least one item must occur again. In this case 2 appears twice and must be returned. There is `n` distinct items, so `+ 1` to the array length means a duplicate must appear next.
2. **Case**: `n = 1` means values are confined to `[1, 1]` meaning the length of the array should be 2 with the items up to 1 - `[1, 1]`. This force to repeat item.
3. **Case (example where the array has far more repetition than the guaranteed minimum)**: `n = 4`, values range `[1, 4]`, the length of the array should be 5. `[3, 3, 3, 3, 3]` - the pigeonhole guarantee ("at least one repeat exists somewhere") is **already satisfied** the moment any repeat shows up, wherever that happens to be. From that point on, pigeonhole itself has nothing left to say about the rest of the array — any remaining constraints on it come from the problem's own rules (exactly one distinct duplicate value, everything else appearing once), not from pigeonhole. Good, that's settled.

#### Step 3 - Brute Force

````python
def find_dupl_number_brute1(nums: list[int]) -> int | None:
    # brute force - nested loop - check every possible pair
    # setup
    # scan input - for loop - [0, len(nums) - 1]
    # scan input - for loop - [1, len(nums)]

    # 1. compare i & j items
    #   i == j: duplicate is found - return item

    for i in range(len(nums) - 1):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                return nums[i]
````

````python
# find_dupl_number_brute1([1, 3, 4, 2, 2])  # 2
# find_dupl_number_brute1([3, 1, 3, 4, 2])  # 3
find_dupl_number_brute1([3, 3, 3, 3, 3])  # 3
````

- **Time**: $O(n^2)$, nested loop, checking every possible pair
- **Space**: $O(1)$ constant, no extra data structures are stored

````python
def find_dupl_number_brute2(nums: list[int]) -> int | None:
    # brute force - hash map - count item occurances
    # setup
    # set a hash map to count item occurances: counts = {}

    # 1. scan input - count elements & store in the hash map
    # 2. scan hash map 'counts' to find a count more than 1 - return its element

    counts = {}

    for i in range(len(nums)):
        counts[nums[i]] = counts.get(nums[i], 0) + 1
        curr_count = counts.get(nums[i], 0)
        if curr_count > 1:
            return nums[i]
````

````python
find_dupl_number_brute2([1, 3, 4, 2, 2])  # 2
# find_dupl_number_brute2([3, 1, 3, 4, 2])  # 3
# find_dupl_number_brute2([3, 3, 3, 3, 3])  # 3
````

- **Time**: $O(n)$ one pass, counting item occurances
- **Space**: $O(n)$ store `n` item occurances in the hash map
