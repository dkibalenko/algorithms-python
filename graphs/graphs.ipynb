## Graph Implementation
### Adjacency List

`node → list of neighbors`

For example, this graph:
```shell
A ── B ── C
│    │
└── D ── E
```
Can be represented as:
```py
{
    "A": ["B", "D"],
    "B": ["A", "C", "D"],
    "C": ["B"],
    "D": ["A", "B", "E"],
    "E": ["D"]
}
```

````python
class Graph:
    def __init__(self):
        self.adj = {}  # adjacency list

    def add_vertex(self, v):
        if v not in self.adj:
            self.adj[v] = []

    def add_edge(self, v1, v2):
        # undirected graph
        self.add_vertex(v1)
        self.add_vertex(v2)

        self.adj[v1].append(v2)
        self.adj[v2].append(v1)

    def __repr__(self):
        return "\n".join(f"{v}: {nbrs}" for v, nbrs in self.adj.items())
````

````python
# Using the Graph
g = Graph()

g.add_edge("A", "B")
g.add_edge("A", "D")
g.add_edge("B", "C")
g.add_edge("B", "D")
g.add_edge("D", "E")

print(g)
````

### Directed Graph Version

````python
def add_edge(self, v1, v2):
    self.add_vertex(v1)
    self.add_vertex(v2)
    self.adj[v1].append(v2)  # only one direction
````
