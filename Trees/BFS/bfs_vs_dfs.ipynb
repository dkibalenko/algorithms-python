### BFS vs DFS

####  BFS ([Link](https://en.wikipedia.org/wiki/Breadth-first_search))
**Pros**

- Shortest Path
- Closer Nodes

**Cons**

- More memory

> Ther rule of thumb: If you have extra info on the location of the target node & you know that the node is likely in the UPPER LEVEL of a tree or graph, then BFS is BETTER.

---

#### DFS ([Link](https://en.wikipedia.org/wiki/Depth-first_search))
**Pros**

- Less Memory
- Does Path Exist? - DFS is built to answer this question

**Cons**

- Can get Slow (When a tree or graph is very deep)

> If you know that the node is likely at the LOWER LEVEL of a tree or graph, perhaps DFS is BETTER.

---

#### BFS:

**Time complexity** is `O(|V|)`, where `|V|` is the number of nodes. You need to traverse all nodes.

**Space complexity** is `O(|V|)` as well - since at worst case you need to hold all vertices in the queue.

#### DFS:

**Time complexity** is again `O(|V|)`, you need to traverse all nodes.

**Space complexity** - depends on the implementation, a recursive implementation can have a `O(h)` space complexity [worst case], where `h` is the maximal depth of your tree.
Using an iterative solution with a stack is actually the same as BFS, just using a stack instead of a queue - so you get both `O(|V|)` time and space complexity.

> Note: the space complexity and time complexity is a bit different for a tree than for a general graphs becase you do not need to maintain a `visited` set for a tree, and `|E| = O(|V|)`, so the `|E|` factor is actually redundant.

---

**Questions You Are Going to Get Asked in an Interview**

1. If you know a solution is not far from the root of the tree: I use BFS since it faster to find the node closer to the root
2. If the tree is very deep and solutions are rare: I use BFS over DFS since if the node is very deep it might be very slow with DFS, but faster with BFS with more memory usage in a queue. Queue for BFS, stack for DFS (or call stack in recursion)
3. If the tree is very wide: - use DFS — BFS must hold an entire level in the queue at once, which explodes for wide trees. DFS only holds the current path, `O(h)` space.
4. If solutions are frequent but located deep in the tree: I use DFS since it's faster to get to the lower part of the tree
5. Determining whether a **path exists between two nodes**:  I use DFS since it's built for this to answer
6. **Finding the shortest path**: I use BFS

### BFS - Iterative level-order traversal

````python
def bfs(self):
    #        9
    #    4      20
    #  1   6 15   170
    curr = self.root
    visited = []  # to store the values of the nodes in the order they were visited
    queue = []  # to keep track of the nodes to visit next (children of the current node)

    if curr:
        queue.append(curr)

    while queue:
        # O(n) shifts all remaining elements left
        curr = queue.pop(0)  # get first node
        visited.append(curr.v)  # add curr node's value

        if curr.left:
            queue.append(curr.left)
        if curr.right:
            queue.append(curr.right)

    return visited
````

### BFS - Recursive level-order traversal

````python
def bfsr(self, queue: list, visited: list = []) -> list:
    if not queue:
        return visited

    curr = queue.pop(0)  # O(n) shifts all remaining elements left
    visited.append(curr.v)
    print(curr)
    if curr.left:
        queue.append(curr.left)
    if curr.right:
        queue.append(curr.right)

    return self.bfsr(queue, visited)
````

### ⏱️ Time complexity

In Python, `list.pop(0)` is **O(n)** because it shifts all remaining elements left.

**Why that matters**

Each node is removed with `pop(0)`, so the total work becomes:
```shell
O(1+2+3+⋯+n)=O(n2)O(1+2+3+⋯+n)=O(n2)
```

So with the current implementation:

- **`bfs` time complexity = **$O(n²)$**
- **`bfsr` time complexity = **$O(n²)$**

**Why it is not just $O(n)$**

Even though you visit each node once, you also pay **extra cost for removing** from the front of a Python list.

### Space complexity
```py
visited = []
```
This stores every node value, so:
- `visited` space = $O(n)$

```py
queue = []
```
- `queue` space = $O(n)$: The queue can hold multiple nodes at once. In the worst case, for a complete tree, the queue may contain roughly half the nodes at one level.

**Total**

So the space usage is:
- `bfs` space = $O(n)$
- `bfsr` spcae = $O(n)$

For `bfsr` there is also recursion depth overhead:
- **Worst-case** recursion depth = $O(n)$ if the tree is **skewed**
- **Balanced tree** recursion depth = $O(log n)$

So the extra call stack makes `bfsr` potentially use:
- $O(n)$ additional stack space in the worst case

### 📌 Summary

| Method                                                                                                                                                                       | What it does                                  |                                                                                                                                                                                                                      Time |                         Space |
| ---------------------------------------------------------------------------------------------------------------------------------------------------------------------------- | --------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------: | ----------------------------: |
| `bfs` | Iterative level-order traversal               | **$O(n^2)$** with current `list.pop(0)`implementation |                      **$O(n)$** |
| `bfsr` | Recursive level-order traversal using a queue | **$O(n²)$** with current `list.pop(0)`implementation | **$O(n)$** plus recursion stack |

### Best fix

Make Time complexity to be $O(n)$, use a `deque` **instead of a Python list**.

````python
from collections import deque

def bfs(self):
    queue = deque([self.root])
    visited = []

    while queue:
        curr = queue.popleft()
        visited.append(curr.v)

        if curr.left:
            queue.append(curr.left)
        if curr.right:
            queue.append(curr.right)

    return visited
````

Then:
- time = $O(n)$
- space = $O(n)$

````python
def bfsr(self, node: Node | None = None, visited: list | None = None) -> list:
    if node is None:
        node = self.root

    if visited is None:
        visited = []

    if node is None:
        return visited

    level = [node]
    return self._bfsr_level(level, visited)

def _bfsr_level(self, level: list[Node], visited: list) -> list:
    if not level:
        return visited

    next_level = []

    for node in level:
        visited.append(node.v)
        print(node)

        if node.left:
            next_level.append(node.left)
        if node.right:
            next_level.append(node.right)

    return self._bfsr_level(next_level, visited)
````

A proper **recursive level-order traversal** by recursing over **levels** instead of popping from the front of a queue.

**What changed**
- `bfsr` now accepts an optional starting node and an optional visited list.
- It builds the current level as a list, visits every node in that level, and recursively processes the next level.

**With the current implementation**:
- Time: O(n)
  - Each node is visited exactly once, and each node’s children are appended once.
- Space: $O(n)$ in the worst case
  - `visited` stores all node values
  - the current `level` / `next_level` lists may hold up to a full level of nodes
  - recursion depth is $O(h)$ for tree height, but the dominant extra space is still the level storage

The old version was effectively worse because `queue.pop(0)` on a Python list shifts elements and can make the runtime `O(n²)`.
