### Core Idea of Binary Search
Binary search repeatedly cuts the search interval in half by comparing the target value to the middle element.

- If the middle element is too big → search the left half.
- If it’s too small → search the right half.

This is why binary search is `O(log n)` — the search space shrinks exponentially.

The main idea of binary search is to keep track of the current **range of reasonable guesses**. In each turn, you choose a guess that divides the set of reasonable guesses into **two ranges** of roughly the same size. If your guess is not correct, then I tell you whether it's too high or too low, and you can eliminate about half of the reasonable guesses.

For example, if the current range of reasonable guesses is 26 to 80, you would guess the halfway point, `(26 + 80) / 2`, or 53. If I then tell you that 53 is too high, you can eliminate all numbers from 53 to 80, leaving 26 to 52 as the new range of reasonable guesses, halving the size of the range.

#### Iterative BS. Step-by-step description of using binary search
1. Let `min = 0` and `max = len(nums) - 1`
2. Compute `mid` as the average of `max` and `min`, rounded down (so that it is an integer)
3. Set `guess = nums[mid]` as your guess
3. If `guess` equals `target`, then stop. You found it! Return `mid`(`target` index)
4. If the guess was **too low**, set `min` to be **one larger** than the guess.
5. If the guess was **too high**, set `max` to be **one smaller** than the guess.
6. Go back to step 2.

```python
def binary_search(nums: list[int], target: int) -> int:
    # set range to search
    low = 0
    high = len(nums) - 1

    while low <= high:  # range of reasonbable guesses
        mid = (low + high) // 2
        guess = nums[mid]

        if guess == target: 
            return mid  # found it!
        
        if guess < target:
            low = mid + 1  # cut off LEFT half
        else:
            high = mid - 1  # cut off RIGHT half

    return None
```

#### Recursive BS

```python
def binary_search_recur(arr: list[int], target: int, left: int, right: int) -> int:
    if left > right:
        return -1  # Base case: target not found

    mid = (left + right) // 2
    guess = arr[mid]

    if guess == target:
        return mid  # Base case: target found
    elif guess < target:
        return binary_search_recur(arr, target, mid + 1, right)  # discard left half
    else:
        return binary_search_recur(arr, target, left, mid - 1)  # discard right half
```
