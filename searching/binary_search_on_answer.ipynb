### Binary Search on Answer - Rope problem
Given an array of integers `ropes` where `ropes[i]` is the length of the rope under `i`, and whole numbers `len` & `k`.

Implement a function `can_cut_pieces`, that defines whether it's possbile to get `k` pieces of a rope with length `len` out of ropes `ropes`. The rope cuttings cannot be combined. The function should return a boolean value.

```shell
can_cut_pieces([10, 5], 3, 5) === false // можливо отримати тільки 3 шматки довжини 3 з першого канату та 1 шматок з другого 

can_cut_pieces([11, 10, 9], 4, 5) === true // можливо отримати по 2 шматки довжиною 4 з кожного канату 
```

This is the classic rope‑cutting problem:

> “Find the maximum piece length such that we can cut at least k pieces.”

**Full Binary Search on Answer** - The classic extension of the rope‑cutting feasibility check.

This is the version used in interviews when they ask:

> “Find the **maximum possible piece length** such that we can cut at least `k` pieces.”

The feasibility function:

````python
def can_cut_pieces(ropes: list[int], length: int, k: int) -> bool:
    pieces = 0
    for rope in ropes:
        pieces += rope // length
        if pieces >= k:
            return True
    return False
````

**Core Idea**

We are searching for the **largest integer length L** such that: we can cut at least `k` pieces of length `L`.

This is a **monotonic predicate**:

- If we can cut pieces of length `L`, then we can also cut pieces of length `L-1`, `L-2`, …

- If we cannot cut pieces of length `L`, then we cannot cut pieces of length `L+1`, `L+2`, …

This monotonicity makes it perfect for binary search.

**Binary Search Boundaries**

- **left** = 1. Minimum possible piece length.

- **right** = `max(ropes)`. Maximum possible piece length (**cannot exceed the longest rope**).

We search for the **maximum feasible length**.

````python
def max_piece_length(ropes: list[int], k: int) -> int:
    left = 1
    right = max(ropes)
    best = 0

    while left <= right:
        mid = (left + right) // 2  # candidate piece length

        if can_cut_pieces(ropes, mid, k):
            best = mid              # mid is feasible → try bigger
            left = mid + 1
        else:
            right = mid - 1         # mid is too big → try smaller

    return best
````

````python
ropes = [11, 10, 9]
k = 5
````

### Iterative approach $O(n)$

You must return:

> `True` if we can cut at least `k` pieces of length `len` from the ropes (no combining).

This is simply: `sum(rope / len) >= k`

````python
def can_cut_pieces(ropes: list[int], length: int, k: int) -> bool:
    pieces = 0
    for rope in ropes:
        pieces += rope // length
        if pieces >= k:
            return True
    return False
````

- Each rope contributes `rope // length` pieces
- We accumulate until we reach `k`
- Early exit makes it efficient

````python
can_cut_pieces([10, 5], 3, 5)
````

````python
can_cut_pieces([11, 10, 9], 4, 5)
````
